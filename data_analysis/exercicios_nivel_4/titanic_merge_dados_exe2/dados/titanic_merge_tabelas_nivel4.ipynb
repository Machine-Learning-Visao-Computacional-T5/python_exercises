## 4 Gabarito comentado
Execute em ordem. Os exercícios 2 a 6 usam a base construída no exercício 1. Os erros intencionais são capturados para não interromper a execução.

### Solução 1 Reconstruir a base analítica

Passos: escolher PassengerId → juntar viagens → juntar resultados → comparar contagem e unicidade. Todas as tabelas foram separadas do mesmo train, então esperamos correspondência completa.

```python
base = passageiros.merge(viagens, on="PassengerId", how="left", validate="one_to_one")
base = base.merge(resultados, on="PassengerId", how="left", validate="one_to_one", indicator=True)
assert len(base) == len(passageiros)
assert base["PassengerId"].is_unique
assert base["_merge"].eq("both").all()
base = base.drop(columns="_merge")
display(base[["PassengerId", "Name", "Pclass", "Embarked", "Survived"]].head())
```

### Solução 2 Enriquecer os portos sem perder passageiros

Passos: manter passageiros à esquerda → ligar códigos com nomes diferentes → usar indicator → inspecionar ausências. O inner exclui quem não encontrou cadastro; isso muda a população analisada.

```python
base_portos = base.merge(portos, left_on="Embarked", right_on="codigo_porto",
                         how="left", validate="many_to_one", indicator=True)
assert len(base_portos) == len(base)
sem_porto = base_portos.loc[base_portos["_merge"].eq("left_only"), ["PassengerId", "Embarked"]]
display(sem_porto)
base_inner = base.merge(portos, left_on="Embarked", right_on="codigo_porto",
                        how="inner", validate="many_to_one")
print("left:", len(base_portos), "inner:", len(base_inner), "sem correspondência:", len(sem_porto))
assert len(base_portos) - len(base_inner) == len(sem_porto)
```

### Solução 3 Bilhetes compartilhados

Passos: agrupar viagens por bilhete → contar passageiros → enriquecer cada passageiro → filtrar → reagrupar. A contagem vale somente para o train. Mesmo Ticket não comprova vínculo familiar. Não some passageiros_no_train após o merge: o total do grupo está repetido em cada passageiro.

```python
bilhetes = (viagens.groupby("Ticket", dropna=False)
            .agg(passageiros_no_train=("PassengerId", "size")).reset_index())
base_bilhetes = base.merge(bilhetes, on="Ticket", how="left", validate="many_to_one")
assert len(base_bilhetes) == len(base)
compartilhados = base_bilhetes.loc[base_bilhetes["passageiros_no_train"] >= 3]
resumo_bilhetes = (compartilhados.groupby("Ticket", dropna=False)
    .agg(passageiros=("PassengerId", "size"), sobreviventes=("Survived", "sum"),
         taxa_sobrevivencia=("Survived", "mean")).reset_index())
resumo_bilhetes["taxa_percentual"] = resumo_bilhetes["taxa_sobrevivencia"] * 100
display(resumo_bilhetes.sort_values(["passageiros", "Ticket"], ascending=[False, True]).head(10))
assert (resumo_bilhetes["sobreviventes"] <= resumo_bilhetes["passageiros"]).all()
```

### Solução 4 Comparar a tarifa com seu grupo

Passos: agrupar por duas colunas → calcular mediana → fazer merge pela chave composta → calcular diferença → filtrar. Fare é a tarifa registrada no dataset; não inferimos preço individual de um bilhete compartilhado. Uma comparação com NaN não seleciona a linha; a mediana ignora valores ausentes.

```python
medianas = (base.groupby(["Pclass", "Sex"], dropna=False)
            .agg(mediana_tarifa=("Fare", "median")).reset_index())
comparacao = base.merge(medianas, on=["Pclass", "Sex"], how="left", validate="many_to_one")
assert len(comparacao) == len(base)
comparacao["diferenca"] = comparacao["Fare"] - comparacao["mediana_tarifa"]
acima = comparacao.loc[comparacao["Fare"] > comparacao["mediana_tarifa"],
                      ["PassengerId", "Pclass", "Sex", "Fare", "mediana_tarifa", "diferenca"]]
display(acima.sort_values("diferenca", ascending=False).head(10))
assert acima["diferenca"].gt(0).all()
```

### Solução 5 Auditar resultados ausentes e chaves órfãs

Passos: introduzir dois problemas controlados → juntar os dois lados → examinar o indicador. NaN representa resultado desconhecido neste cruzamento; preencher com zero mudaria o significado para morte observada.

```python
id_removido = resultados.iloc[0]["PassengerId"]
id_orfao = int(base["PassengerId"].max()) + 1
resultados_auditoria = pd.concat([
    resultados.loc[resultados["PassengerId"] != id_removido],
    pd.DataFrame({"PassengerId": [id_orfao], "Survived": [1]})
], ignore_index=True)
auditoria = passageiros.merge(resultados_auditoria, on="PassengerId", how="outer",
                               validate="one_to_one", indicator=True)
display(auditoria.loc[auditoria["_merge"] != "both", ["PassengerId", "Name", "Survived", "_merge"]])
assert auditoria["_merge"].eq("left_only").sum() == 1
assert auditoria["_merge"].eq("right_only").sum() == 1
assert auditoria.loc[auditoria["_merge"].eq("left_only"), "Survived"].isna().all()
```

### Solução 6 Investigar um merge que infla contagens

Passos: testar cardinalidade → observar multiplicação → identificar IDs repetidos → corrigir a origem. Se duas linhas tiverem o mesmo código e nomes conflitantes, drop_duplicates não resolve: é necessário decidir qual registro é válido.

```python
portos_duplicados = pd.concat([portos, portos.iloc[[0]]], ignore_index=True)
try:
    base.merge(portos_duplicados, left_on="Embarked", right_on="codigo_porto",
               how="left", validate="many_to_one")
except pd.errors.MergeError as erro:
    print("Erro esperado de cardinalidade:", erro)
else:
    raise AssertionError("A validação deveria detectar a chave duplicada.")
inflada = base.merge(portos_duplicados, left_on="Embarked", right_on="codigo_porto", how="left")
print("Linhas originais:", len(base), "após merge:", len(inflada), "inflação:", len(inflada)-len(base))
display(inflada.loc[inflada["PassengerId"].duplicated(keep=False), ["PassengerId", "Embarked", "nome_porto"]].head(10))
# Neste caso a duplicação é exata e foi criada artificialmente.
portos_corrigidos = portos_duplicados.drop_duplicates()
assert portos_corrigidos["codigo_porto"].is_unique
corrigida = base.merge(portos_corrigidos, left_on="Embarked", right_on="codigo_porto",
                       how="left", validate="many_to_one")
assert len(corrigida) == len(base)
assert corrigida["PassengerId"].is_unique
assert len(inflada)-len(base) == base["Embarked"].eq(portos.iloc[0]["codigo_porto"]).sum()
```
