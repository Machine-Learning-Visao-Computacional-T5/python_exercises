## Atividade 3 — sobrevivência por sexo e classe

**Enunciado:** construir passageiros, sobreviventes e taxa percentual por sexo e classe, ordenar por taxa e conferir o total.

**Ideia:** dividir em grupos → calcular dentro de cada grupo → comparar os resumos.

### 3.1. Entender a média de 0 e 1

**O que queremos:** Ver por que mean pode medir proporção.

**Explique antes de executar:**

Nos valores [0, 1, 1, 0], há dois sobreviventes em quatro pessoas. Soma = 2; média = 2/4 = 0,5. Multiplicar por 100 dá 50%. Isso funciona porque o indicador usa 0 e 1.

```python
exemplo = pd.Series([0, 1, 1, 0])
print("Soma:", exemplo.sum())
print("Média:", exemplo.mean())
print("Percentual:", exemplo.mean() * 100)
```

**O que observar e dizer na gravação:** Uma média de 0,5 aqui não é meio passageiro: é uma proporção de sobrevivência.

### 3.2. Agrupar e resumir

**O que queremos:** Produzir uma linha por combinação de sexo e classe.

**Explique antes de executar:**

`groupby(["Sex", "Pclass"])` define os grupos. `.agg(...)` calcula estatísticas. Na escrita `passageiros=("PassengerId", "size")`, passageiros é o nome de saída; PassengerId é a coluna de entrada; size conta as linhas. sum soma o indicador; mean calcula a proporção.

```python
resumo = titanic.groupby(["Sex", "Pclass"]).agg(
    passageiros=("PassengerId", "size"),
    sobreviventes=("Survived", "sum"),
    taxa=("Survived", "mean")
)
resumo
```

**O que observar e dizer na gravação:** Temos seis grupos. Os nomes dos grupos aparecem no índice; taxa ainda está entre 0 e 1.

### 3.3. Transformar o índice em colunas

**O que queremos:** Deixar o resumo com colunas comuns.

**Explique antes de executar:**

`.reset_index()` transforma os identificadores dos grupos em colunas e cria um índice simples. Não altera a quantidade de passageiros.

```python
resumo = resumo.reset_index()
resumo
```

**O que observar e dizer na gravação:** Sex e Pclass agora podem ser acessadas como outras colunas.

### 3.4. Percentual e ordenação

**O que queremos:** Comparar grupos pela taxa.

**Explique antes de executar:**

Crie taxa_percentual multiplicando a proporção por 100. Ordene pela nova coluna em ordem decrescente.

```python
resumo["taxa_percentual"] = resumo["taxa"] * 100
resumo.sort_values("taxa_percentual", ascending=False)
```

**O que observar e dizer na gravação:** Neste arquivo, mulheres da classe 1 têm aproximadamente 96,81%; homens da classe 3, 13,54%. São associações observadas, sem demonstração de causa.

### 3.5. Conferir o total

**O que queremos:** Verificar que as contagens somam o tamanho da base.

**Explique antes de executar:**

`.sum()` agora soma uma coluna numérica do resumo. Comparamos ao total original usando ==.

```python
print("Total nos grupos:", resumo["passageiros"].sum())
print("Total na base:", len(titanic))
print("Conferência:", resumo["passageiros"].sum() == len(titanic))
```

**O que observar e dizer na gravação:** 891 em ambos; True na conferência. groupby pode excluir chaves ausentes por padrão; aqui Sex e Pclass não têm ausências.

## Atividade 4 — investigar e preencher idades ausentes

**Enunciado:** comparar registros com idade conhecida e ausente; preencher numa cópia usando mediana por sexo e classe, manter Age original e marcar imputações.

**Imputar** significa preencher uma ausência com um valor estimado por uma regra. Não significa recuperar a idade verdadeira. **Mediana** é o valor central dos valores ordenados; com quantidade par, é a média dos dois centrais.

### 4.1. Criar cópia e indicador

**O que queremos:** Separar o experimento de preenchimento da tabela original.

**Explique antes de executar:**

`.copy()` cria outra tabela. `.isna()` na coluna Age devolve a marcação de ausência. Guardamos essa marcação antes de preencher.

```python
idades = titanic.copy()
idades["idade_ausente"] = idades["Age"].isna()
idades[["Age", "idade_ausente"]].head(10)
```

**O que observar e dizer na gravação:** True significa que a idade original não estava disponível; False significa que havia idade registrada.

### 4.2. Comparar os dois grupos

**O que queremos:** Ver contagem e taxa segundo presença de idade.

**Explique antes de executar:**

Agrupe por idade_ausente. False reúne idades conhecidas; True reúne desconhecidas. A taxa considera Survived, que está preenchida neste arquivo.

```python
comparacao_idades = idades.groupby("idade_ausente").agg(
    passageiros=("PassengerId", "size"),
    taxa=("Survived", "mean")
)
comparacao_idades["taxa_percentual"] = comparacao_idades["taxa"] * 100
comparacao_idades
```

**O que observar e dizer na gravação:** 714 idades conhecidas, com taxa de aproximadamente 40,62%; 177 ausentes, com 29,38%. Isso descreve grupos, não prova que ausência de idade afeta sobrevivência.

### 4.3. Calcular medianas dos grupos

**O que queremos:** Conhecer o valor que será usado para cada grupo.

**Explique antes de executar:**

Agrupe por Sex e Pclass, selecione Age e calcule median(). A mediana ignora idades ausentes.

```python
idades.groupby(["Sex", "Pclass"])["Age"].median()
```

**O que observar e dizer na gravação:** As seis medianas são: mulheres classes 1/2/3 = 35, 28, 21,5; homens classes 1/2/3 = 40, 30, 25.

### 4.4. Alinhar a mediana a cada passageiro

**O que queremos:** Obter uma mediana de grupo para cada linha da base.

**Explique antes de executar:**

`.transform("median")` calcula a mediana dentro do grupo e repete o resultado nas linhas daquele grupo. Diferente da agregação, mantém 891 posições alinhadas aos passageiros.

```python
mediana_grupo = idades.groupby(["Sex", "Pclass"])["Age"].transform("median")
mediana_grupo.head(10)
```

**O que observar e dizer na gravação:** Não temos seis linhas, mas uma sugestão por passageiro. Esse alinhamento permite preencher sem merge.

### 4.5. Preencher numa nova coluna

**O que queremos:** Substituir apenas as ausências, preservando Age.

**Explique antes de executar:**

`.fillna(mediana_grupo)` mantém valores existentes e usa a mediana correspondente somente onde havia ausência. Guardamos em Age_imputada.

```python
idades["Age_imputada"] = idades["Age"].fillna(mediana_grupo)
idades.loc[idades["idade_ausente"], ["PassengerId", "Age", "Age_imputada"]].head(10)
```

**O que observar e dizer na gravação:** Age continua ausente nas linhas mostradas; Age_imputada tem estimativas. idade_ausente também registra quais valores foram imputados neste arquivo.

### 4.6. Conferir o preenchimento

**O que queremos:** Ver nulos restantes e preservação do original.

**Explique antes de executar:**

`.equals(...)` compara a Series original à da cópia. Se um grupo não tivesse idades conhecidas, sua mediana seria ausente e o preenchimento não resolveria tudo.

```python
print("Nulos após preencher:", idades["Age_imputada"].isna().sum())
print("Age original preservada:", idades["Age"].equals(titanic["Age"]))
```

**O que observar e dizer na gravação:** Neste CSV: 0 nulos restantes e True para preservação. Em modelagem, estatísticas de imputação devem ser aprendidas somente nos dados de treino de cada divisão; aqui a atividade é descritiva.

## Atividade 5 — familiares a bordo

**Enunciado:** criar tamanho_familia = SibSp + Parch + 1 e viajava_sozinho; comparar passageiros, sobreviventes e taxas.

A regra usa familiares registrados nos campos. Não identifica amigos nem todos os acompanhantes; bilhete compartilhado não entra nesta definição.

### 5.1. Calcular tamanho de família

**O que queremos:** Somar familiares e a própria pessoa.

**Explique antes de executar:**

As somas entre colunas são feitas linha a linha. `+ 1` inclui o próprio passageiro.

```python
familias = titanic.copy()
familias["tamanho_familia"] = familias["SibSp"] + familias["Parch"] + 1
familias[["SibSp", "Parch", "tamanho_familia"]].head(20)
```

**O que observar e dizer na gravação:** SibSp 0 e Parch 0 produzem tamanho 1. Não somamos todos os passageiros em uma única família.

### 5.2. Marcar ausência de familiares

**O que queremos:** Identificar tamanho_familia igual a 1.

**Explique antes de executar:**

`.eq(1)` faz a mesma comparação que `== 1`. O nome viajava_sozinho é operacional: significa sem familiares registrados por essa regra.

```python
familias["viajava_sozinho"] = familias["tamanho_familia"].eq(1)
familias[["tamanho_familia", "viajava_sozinho"]].head()
```

**O que observar e dizer na gravação:** True marca ausência de familiares registrados; não comprova que não havia companhia.

### 5.3. Comparar os grupos

**O que queremos:** Resumir sobrevivência segundo essa definição.

**Explique antes de executar:**

Reutilize groupby e agg. Adicionamos percentual para facilitar a leitura; a proporção continua em taxa.

```python
resumo_familias = familias.groupby("viajava_sozinho").agg(
    passageiros=("PassengerId", "size"),
    sobreviventes=("Survived", "sum"),
    taxa=("Survived", "mean")
)
resumo_familias["taxa_percentual"] = resumo_familias["taxa"] * 100
resumo_familias
```

**O que observar e dizer na gravação:** Sem familiares: 537 passageiros, 163 sobreviventes, aproximadamente 30,35%. Com familiares: 354, 179, aproximadamente 50,56%. Não conclua que companhia causou sobrevivência.

## Atividade 6 — faixas etárias

### 6.1. Criar intervalos

**O que queremos:** Converter idades numéricas em categorias.

**Explique antes de executar:**

`pd.cut` recebe a coluna, limites em bins e nomes em labels. `right=False` inclui o limite esquerdo e exclui o direito: [12,18), por exemplo. `float("inf")` é infinito positivo, para não impor teto à última faixa.

```python
faixas = titanic.copy()
faixas["faixa_etaria"] = pd.cut(
    faixas["Age"],
    bins=[0, 12, 18, 60, float("inf")],
    right=False,
    labels=["Menos de 12", "12 a menos de 18", "18 a menos de 60", "60 ou mais"]
)
faixas[["Age", "faixa_etaria"]].head(10)
```

**O que observar e dizer na gravação:** Idades ausentes continuam sem categoria. Não usamos Age_imputada, para não misturar estimativa com observação.

### 6.2. Registrar a categoria de ausência

**O que queremos:** Dar um nome visível às idades desconhecidas.

**Explique antes de executar:**

cut produz uma coluna categórica. Antes de preencher com um novo rótulo, adicionamos esse rótulo às categorias permitidas via `.cat.add_categories`. Depois fillna usa o rótulo.

```python
faixas["faixa_etaria"] = faixas["faixa_etaria"].cat.add_categories("Não informado")
faixas["faixa_etaria"] = faixas["faixa_etaria"].fillna("Não informado")
faixas.loc[faixas["Age"].isna(), ["Age", "faixa_etaria"]].head()
```

**O que observar e dizer na gravação:** A idade original segue ausente. Somente a coluna de faixa ganha o texto Não informado.

### 6.3. Comparar faixas

**O que queremos:** Contar e calcular taxa por categoria.

**Explique antes de executar:**

`observed=True` pede somente categorias que aparecem nos dados. Com a categoria Não informado, os 177 passageiros sem idade são incluídos no resumo.

```python
resumo_faixas = faixas.groupby("faixa_etaria", observed=True).agg(
    passageiros=("PassengerId", "size"),
    taxa=("Survived", "mean")
)
resumo_faixas["taxa_percentual"] = resumo_faixas["taxa"] * 100
resumo_faixas
```

**O que observar e dizer na gravação:** Contagens: 68, 45, 575, 26 e 177, respectivamente. Somam 891. Compare os tamanhos antes de interpretar diferenças de taxas.

## Atividade 7 — relatório final

**Enunciado:** por classe e presença de familiares, produzir passageiros, sobreviventes, taxa percentual, mediana de tarifa e idade e número de idades conhecidas. Destacar grupos com pelo menos 20 registros e taxa acima da geral.

**Plano:** criar a chave de familiares → resumir → converter taxa → calcular referência geral → filtrar → escrever observações.

### 7.1. Preparar a chave de familiares

**O que queremos:** Criar a definição usada no relatório.

**Explique antes de executar:**

Somar SibSp e Parch e comparar a zero equivale a verificar tamanho_familia igual a 1. Usamos outra cópia para o relatório.

```python
relatorio_base = titanic.copy()
relatorio_base["viajava_sozinho"] = (relatorio_base["SibSp"] + relatorio_base["Parch"]).eq(0)
relatorio_base[["Pclass", "viajava_sozinho"]].head()
```

**O que observar e dizer na gravação:** Temos as duas chaves para os grupos: classe e indicação de ausência de familiares.

### 7.2. Calcular todas as métricas

**O que queremos:** Formar uma tabela com uma linha por grupo.

**Explique antes de executar:**

`size` conta todas as linhas; `count` em Age conta somente valores conhecidos. `median` em Age ignora ausências. Mediana de tarifa resume a tarifa registrada; não demonstra quanto cada pessoa efetivamente pagou em todas as situações.

```python
relatorio = relatorio_base.groupby(["Pclass", "viajava_sozinho"]).agg(
    passageiros=("PassengerId", "size"),
    sobreviventes=("Survived", "sum"),
    taxa=("Survived", "mean"),
    mediana_tarifa=("Fare", "median"),
    mediana_idade=("Age", "median"),
    idades_conhecidas=("Age", "count")
).reset_index()
relatorio["taxa_percentual"] = relatorio["taxa"] * 100
relatorio
```

**O que observar e dizer na gravação:** Se passageiros e idades_conhecidas diferem, a mediana de idade foi calculada sobre um subconjunto. Existem seis grupos neste arquivo.

### 7.3. Calcular taxa geral

**O que queremos:** Criar a referência para comparação.

**Explique antes de executar:**

`Survived.mean()` calcula sobreviventes dividido pelo total neste CSV. Guardamos como proporção e mostramos em percentual.

```python
taxa_geral = titanic["Survived"].mean()
print("Taxa geral (%):", round(taxa_geral * 100, 2))
```

**O que observar e dizer na gravação:** 342 / 891: aproximadamente 38,38%. round(..., 2) arredonda a exibição para duas casas.

### 7.4. Destacar grupos

**O que queremos:** Escolher grupos suficientemente grandes e acima da taxa geral.

**Explique antes de executar:**

`>= 20` inclui grupos com exatamente 20. `> taxa_geral` exige taxa estritamente maior. Use proporção com proporção, sem comparar taxa_percentual com taxa_geral.

```python
destaque = (relatorio["passageiros"] >= 20) & (relatorio["taxa"] > taxa_geral)
grupos_destacados = relatorio.loc[destaque]
grupos_destacados
```

**O que observar e dizer na gravação:** São três grupos: classe 1 com familiares, classe 1 sem familiares e classe 2 com familiares. O corte 20 vem do exercício, não de uma garantia estatística.

### 7.5. Conferir total do relatório

**O que queremos:** Garantir coerência das contagens.

**Explique antes de executar:**

A soma das contagens dos grupos deve recuperar a base. Essa checagem é uma evidência de consistência, não uma prova de qualidade completa.

```python
print("Passageiros no relatório:", relatorio["passageiros"].sum())
print("Sobreviventes no relatório:", relatorio["sobreviventes"].sum())
```

**O que observar e dizer na gravação:** 891 passageiros e 342 sobreviventes.

### 7.6. Resposta escrita — observações e limitação

**Observação 1:** no arquivo fornecido, a classe 1 apresenta taxa de sobrevivência acima da taxa geral tanto entre pessoas com familiares registrados quanto entre pessoas sem familiares registrados. O grupo com familiares tem taxa aproximada de 72,90%; o grupo sem familiares, 53,21%.

**Observação 2:** na classe 2, o grupo com familiares registrados apresenta aproximadamente 63,75%, contra 34,62% no grupo sem familiares registrados. O resumo descreve uma diferença, mas ainda não separa possíveis efeitos de idade, sexo e outras características.

**Limitação:** o arquivo contém 891 registros, não todas as pessoas a bordo; 177 idades estão ausentes, e a definição de companhia usa apenas SibSp e Parch. A análise é descritiva e não demonstra que classe ou familiares causaram a sobrevivência.
