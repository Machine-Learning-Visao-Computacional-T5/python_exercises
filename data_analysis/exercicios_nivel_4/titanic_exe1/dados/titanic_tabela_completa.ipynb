### Solução 1 Conhecer os dados

O denominador dos percentuais é o número de linhas.

```python
print(titanic.shape)
print(titanic.columns.tolist())
print(titanic.dtypes)
nulos = titanic.isna().sum().to_frame("quantidade")
nulos["percentual"] = nulos["quantidade"] / len(titanic) * 100
display(nulos.sort_values("quantidade", ascending=False))
display(nulos.sort_values("quantidade", ascending=False).head(3))
```

### Solução 2 Filtros combinados

Combine condições com & e use parênteses; isin seleciona duas classes.

```python
filtro = ((titanic["Sex"] == "female") & titanic["Pclass"].isin([1, 2])
          & (titanic["Age"] >= 18) & (titanic["Fare"] > 30))
selecionadas = titanic.loc[filtro, ["PassengerId", "Name", "Age", "Pclass", "Fare"]]
display(selecionadas)
print("Selecionadas:", len(selecionadas))
assert selecionadas["Age"].ge(18).all()
```

NaN não atende à comparação Age >= 18; as idades desconhecidas ficam fora do filtro. Não podemos afirmar que essas pessoas são menores de idade.

### Solução 3 Sobrevivência por sexo e classe

A média de uma coluna 0/1 é a proporção de valores 1.

```python
resumo = titanic.groupby(["Sex", "Pclass"]).agg(
    passageiros=("PassengerId", "size"), sobreviventes=("Survived", "sum"),
    taxa=("Survived", "mean")).reset_index()
resumo["taxa_percentual"] = resumo["taxa"] * 100
display(resumo.sort_values("taxa_percentual", ascending=False))
assert resumo["passageiros"].sum() == len(titanic)
```

### Solução 4 Investigar idades ausentes

transform devolve um valor alinhado a cada linha, sem merge. Imputação não recupera a idade real.

```python
idades = titanic.copy()
idades["idade_ausente"] = idades["Age"].isna()
display(idades.groupby("idade_ausente").agg(
    passageiros=("PassengerId", "size"), taxa=("Survived", "mean")))
mediana_grupo = idades.groupby(["Sex", "Pclass"])["Age"].transform("median")
idades["Age_imputada"] = idades["Age"].fillna(mediana_grupo)
print("Nulos restantes:", idades["Age_imputada"].isna().sum())
assert idades["Age"].equals(titanic["Age"])
display(idades.loc[idades["idade_ausente"], ["PassengerId", "Age", "Age_imputada"]].head())
```

Se um grupo não tiver nenhuma idade conhecida, sua mediana também será ausente. Verifique nulos restantes. Para modelagem, calcule estatísticas de imputação apenas no conjunto de treino de cada divisão; aqui a atividade é descritiva.

### Solução 5 Familiares a bordo

Uma pessoa conta a si mesma; Ticket compartilhado não é usado para definir família.

```python
familias = titanic.copy()
familias["tamanho_familia"] = familias["SibSp"] + familias["Parch"] + 1
familias["viajava_sozinho"] = familias["tamanho_familia"].eq(1)
display(familias.groupby("viajava_sozinho").agg(
    passageiros=("PassengerId", "size"), sobreviventes=("Survived", "sum"),
    taxa=("Survived", "mean")))
assert familias["tamanho_familia"].ge(1).all()
```

### Solução 6 Faixas etárias

Defina os limites antes do código. Não use idades imputadas como se fossem observadas.

```python
faixas = titanic.copy()
faixas["faixa_etaria"] = pd.cut(faixas["Age"],
    bins=[0, 12, 18, 60, float("inf")], right=False,
    labels=["Menos de 12", "12 a menos de 18", "18 a menos de 60", "60 ou mais"])
faixas["faixa_etaria"] = faixas["faixa_etaria"].cat.add_categories("Não informado").fillna("Não informado")
resumo_faixas = faixas.groupby("faixa_etaria", observed=True).agg(
    passageiros=("PassengerId", "size"), taxa=("Survived", "mean"))
display(resumo_faixas)
assert resumo_faixas["passageiros"].sum() == len(titanic)
```

### Solução 7 Relatório final

O tamanho dos grupos importa; a mediana de idade ignora nulos. Comparações descrevem associação, não causalidade.

```python
relatorio_base = titanic.copy()
relatorio_base["viajava_sozinho"] = (relatorio_base["SibSp"] + relatorio_base["Parch"]).eq(0)
relatorio = relatorio_base.groupby(["Pclass", "viajava_sozinho"]).agg(
    passageiros=("PassengerId", "size"), sobreviventes=("Survived", "sum"),
    taxa=("Survived", "mean"), mediana_tarifa=("Fare", "median"),
    mediana_idade=("Age", "median"), idades_conhecidas=("Age", "count")).reset_index()
relatorio["taxa_percentual"] = relatorio["taxa"] * 100
taxa_geral = titanic["Survived"].mean()
display(relatorio)
print("Taxa geral (%):", taxa_geral * 100)
display(relatorio.loc[(relatorio["passageiros"] >= 20) & (relatorio["taxa"] > taxa_geral)])
assert relatorio["passageiros"].sum() == len(titanic)
```
