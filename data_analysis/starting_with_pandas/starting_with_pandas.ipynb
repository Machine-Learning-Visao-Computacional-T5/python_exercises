## Antes do pandas: função, método, parâmetro e argumento

Execute os exemplos desta seção de cima para baixo. Eles usam apenas Python e não precisam de pandas.

### 1. Função: uma tarefa com nome

Uma **função** é um bloco de código que realiza uma tarefa. Usamos funções para reaproveitar uma tarefa sem escrever suas instruções novamente.

Criar uma função e chamar uma função são passos diferentes. Ao executar `def`, Python registra a função. O corpo será executado quando fizermos uma chamada.

Na próxima célula:

- `def` indica que estamos definindo uma função.
- `dobrar` é o nome escolhido para ela.
- `numero` é um **parâmetro**: o nome que receberá um valor na chamada.
- `:` inicia o bloco da função.
- Os espaços antes de `return` indicam que essa linha pertence à função.
- `* 2` multiplica o valor por dois.
- `return` devolve o resultado para quem chamou a função. Ele não exibe o resultado na tela.


````python
def dobrar(numero):
    return numero * 2
````

### 2. Chamada: executar a função e guardar o retorno

Na próxima célula, Python calcula primeiro o lado direito de `=`:

1. Chama `dobrar(5)`.
2. Dentro da função, o parâmetro `numero` recebe `5`.
3. Calcula `5 * 2` e devolve `10`.
4. Guarda esse retorno na variável `resultado`.
5. `print(resultado)` mostra o valor guardado.

`dobrar` sem parênteses se refere à função. `dobrar(5)` chama a função. A saída esperada é `10`.


````python
resultado = dobrar(5)
print(resultado)
````

### 3. Parâmetro e argumento

Um **argumento** é o valor enviado para uma função ou método em uma chamada.

| Conceito | Exemplo | Papel |
|---|---|---|
| Parâmetro | `numero` em `def dobrar(numero)` | Nome que recebe o valor dentro da função |
| Argumento | `5` em `dobrar(5)` | Valor enviado na chamada |
| Retorno | `10` | Resultado devolvido pela função |
| Variável que guarda o retorno | `resultado` | Nome usado para acessar esse resultado depois |

Uma função pode receber mais de um argumento. No exemplo abaixo, `20` vai para `preco` e `3` vai para `quantidade`. A vírgula separa os argumentos.


````python
def calcular_total(preco, quantidade):
    return preco * quantidade

total = calcular_total(20, 3)
print(total)
````

A saída é `60`. Podemos também informar os nomes dos parâmetros na chamada, usando **argumentos nomeados**:

`preco=20` configura o parâmetro desta chamada; não cria uma variável global chamada `preco`. Neste caso, os nomes permitem escrever os argumentos em outra ordem.


````python
total_nomeado = calcular_total(quantidade=3, preco=20)
print(total_nomeado)
````

### 4. Método: uma função associada a um objeto

Um **objeto** é um valor com um tipo, informações e operações disponíveis. Textos, listas e tabelas são exemplos de objetos.

Um **método** é uma função associada a um objeto. O texto guardado em `nome` possui o método `upper`, que retorna uma versão com letras maiúsculas.

Leia `nome.upper()` assim: “no texto guardado em `nome`, execute o método `upper`”.

- `nome` é a variável que contém o texto.
- `.` acessa algo disponível nesse objeto.
- `upper` é o nome do método.
- `()` fazem a chamada. Esse método não precisa receber argumentos explícitos.
- `nome_maiusculo` guarda o texto retornado.

A saída é `ANA`. O texto original em `nome` continua sendo `Ana`, porque `upper()` retorna outro texto.


````python
nome = "Ana"
nome_maiusculo = nome.upper()
print(nome_maiusculo)
print(nome)
````

Um método também pode receber argumentos. `replace("Ana", "Carla")` recebe dois textos: o trecho a procurar e o texto que o substituirá. A saída esperada é `Olá, Carla`.


````python
mensagem = "Olá, Ana"
nova_mensagem = mensagem.replace("Ana", "Carla")
print(nova_mensagem)
````

### 5. Atributo: uma informação disponível no objeto

Um **atributo** fornece uma informação do objeto. Nos exemplos do pandas, `alunos.shape` fornece a quantidade de linhas e colunas. Não usamos parênteses para consultar essa informação.

Essas expressões são para leitura neste momento. Execute os exemplos com `alunos` depois que a tabela for criada no notebook. O CSV só pode ser lido se o arquivo existir.
