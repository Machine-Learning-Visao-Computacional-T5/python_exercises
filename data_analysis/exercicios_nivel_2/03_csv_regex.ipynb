**Conceito novo:** regex, ou expressão regular, descreve um padrão de texto. Vamos verificar códigos no formato PED- seguido de exatamente três dígitos e retirar pontuação dos telefones. Esta verificação é de formato; não comprova que um pedido ou telefone existe.

| Símbolo | Significado |
|---|---|
| `PED-` | Texto literal |
| `[0-9]` | Um dígito de 0 a 9 |
| `{3}` | Exatamente três repetições |
| `\D` | Um caractere que não é dígito |

`r"..."` é uma string bruta, útil para escrever barras invertidas nos padrões. `str.fullmatch` verifica o texto inteiro; `str.replace` substitui partes que combinam com o padrão.

## Parte 2 — resolução comentada

Compare com suas tentativas. Leia como pensar, execute o código e explique o resultado com suas palavras. Comece executando a preparação novamente.

```python
import pandas as pd

df = pd.read_csv("pedidos_regex.csv", dtype={"telefone": "string"}, encoding="utf-8")
df
```

### Resolução 1

**Pedido:** Mostre as primeiras linhas e o tamanho da tabela.

**Como pensar:** Começamos olhando os dados antes de escrever o padrão.

```python
print(df.shape)
df.head()
```

**Resultado explicado:** 6 pedidos e 4 colunas.

### Resolução 2

**Pedido:** Inspecione as colunas e os tipos.

**Como pensar:** Telefone é informação textual, não uma quantidade. Na leitura pedimos dtype string para essa coluna.

```python
print(df.columns)
df.info()
```

**Resultado explicado:** cliente, codigo, telefone e valor. Não devemos calcular média de telefone.

### Resolução 3

**Pedido:** Mostre nome e código usando loc.

**Como pensar:** Selecionamos duas colunas para examinar os exemplos antes da regex.

```python
df.loc[:, ["cliente", "codigo"]]
```

**Resultado explicado:** XYZ-003 e PED-ABC fogem do formato pretendido.

### Resolução 4

**Pedido:** Calcule a média dos valores dos pedidos.

**Como pensar:** Selecionamos a coluna numérica e calculamos mean.

```python
df["valor"].mean()
```

**Resultado explicado:** A soma é 660; a média é 110 reais.

### Resolução 5

**Pedido:** Verifique quais códigos seguem PED- e três dígitos.

**Como pensar:** PED- é fixo; [0-9] é um dígito; {3} exige três. fullmatch exige correspondência do texto inteiro. na=False trata uma eventual ausência como não correspondente.

```python
padrao = r"PED-[0-9]{3}"
codigo_valido = df["codigo"].str.fullmatch(padrao, na=False)
print(codigo_valido)
```

**Resultado explicado:** True para Ana, Bruno, Diego e Felipe. False para Carla e Elisa.

### Resolução 6

**Pedido:** Mostre os pedidos com códigos fora do padrão.

**Como pensar:** O símbolo ~ inverte True e False nessa Series booleana.

```python
df.loc[~codigo_valido, ["cliente", "codigo"]]
```

**Resultado explicado:** Carla: XYZ-003; Elisa: PED-ABC.

### Resolução 7

**Pedido:** Crie uma cópia e uma coluna telefone_limpo só com dígitos.

**Como pensar:** \D encontra caracteres que não são dígitos. Substituímos cada ocorrência por texto vazio. regex=True ativa a interpretação do padrão. Preservamos a coluna original.

```python
df_limpo = df.copy()
df_limpo["telefone_limpo"] = df_limpo["telefone"].str.replace(r"\D", "", regex=True)
df_limpo.loc[:, ["telefone", "telefone_limpo"]]
```

**Resultado explicado:** Ana fica com 41900000001; Bruno com 41900000002. Limpar a pontuação não verifica se o telefone é real.

### Resolução 8

**Pedido:** Mostre os clientes cujo telefone limpo começa com 41.

**Como pensar:** Depois da limpeza, startswith verifica o prefixo. Para um prefixo literal simples não é necessário usar regex.

```python
df_limpo.loc[df_limpo["telefone_limpo"].str.startswith("41", na=False), "cliente"]
```

**Resultado explicado:** Ana, Bruno, Carla e Felipe.

### Resolução 9

**Pedido:** Exporte os dados com a nova coluna para CSV.

**Como pensar:** Usamos outro nome para criar uma saída com os textos originais e a coluna limpa.

```python
df_limpo.to_csv("pedidos_limpos.csv", index=False, encoding="utf-8")
```

**Resultado explicado:** pedidos_limpos.csv é criado. No Colab, use o menu do arquivo para baixar.

## Lembretes

| Recurso | Significado |
|---|---|
| `df["coluna"]` | Seleciona uma coluna |
| `df[condicao]` | Seleciona as linhas que atendem à condição |
| `df.loc[linhas, colunas]` | Seleciona por rótulos ou condições |
| `df.iloc[linhas, colunas]` | Seleciona por posições começando em zero |
| `:` sozinho | Todas as linhas ou colunas |
| `0:3` no iloc | Posições 0, 1 e 2; exclui a posição 3 |
| `==` | Compara valores |
| `=` | Atribui um valor |

Selecionar ou mostrar um resultado não significa apagar as outras informações. A alteração de valores é feita explicitamente, como na questão 9.
