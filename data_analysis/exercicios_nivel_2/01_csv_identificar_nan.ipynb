## Parte 2 — resolução comentada

Compare com suas tentativas. Leia como pensar, execute o código e explique o resultado com suas palavras. Comece executando a preparação novamente.

```python
import pandas as pd

df = pd.read_csv("roupas_nan.csv", encoding="utf-8")
df
```

### Resolução 1

**Pedido:** Mostre as três primeiras linhas.

**Como pensar:** Primeiro visualizamos o arquivo que acabamos de abrir.

```python
df.head(3)
```

**Resultado explicado:** Camiseta, Calça e Vestido aparecem; o preço do Vestido está ausente.

### Resolução 2

**Pedido:** Veja o tamanho, os nomes das colunas e um resumo da tabela.

**Como pensar:** shape mostra linhas e colunas. columns lista os nomes. info informa tipos e quantidades preenchidas.

```python
print(df.shape)
print(df.columns)
df.info()
```

**Resultado explicado:** São 6 linhas e 3 colunas; preco e estoque têm 5 valores preenchidos cada.

### Resolução 3

**Pedido:** Mostre somente os preços.

**Como pensar:** Selecionamos a coluna antes de fazer cálculos.

```python
df["preco"]
```

**Resultado explicado:** O Vestido não tem preço registrado. Não podemos concluir que ele é gratuito.

### Resolução 4

**Pedido:** Calcule a média dos preços registrados.

**Como pensar:** mean ignora os valores ausentes por padrão.

```python
df["preco"].mean()
```

**Resultado explicado:** A soma dos cinco preços é 510; a média é 102. Não é a média de todos os produtos, pois falta um preço.

### Resolução 5

**Pedido:** Identifique os valores ausentes com isna.

**Como pensar:** isna verifica cada célula; True indica ausência.

```python
df.isna()
```

**Resultado explicado:** True aparece no preço do Vestido e no estoque da Jaqueta.

### Resolução 6

**Pedido:** Conte os valores ausentes de cada coluna.

**Como pensar:** Na soma, True conta como 1 e False como 0; a soma padrão é por coluna.

```python
df.isna().sum()
```

**Resultado explicado:** peca: 0; preco: 1; estoque: 1.

### Resolução 7

**Pedido:** Mostre todas as informações das peças sem preço.

**Como pensar:** O resultado booleano de isna seleciona as linhas ausentes.

```python
df[df["preco"].isna()]
```

**Resultado explicado:** Somente Vestido aparece.

### Resolução 8

**Pedido:** Mostre os nomes das peças com preço registrado usando loc.

**Como pensar:** notna identifica preenchidos; loc seleciona essas linhas e a coluna peca.

```python
df.loc[df["preco"].notna(), "peca"]
```

**Resultado explicado:** Camiseta, Calça, Jaqueta, Shorts e Saia.

### Resolução 9

**Pedido:** Compare estoque zero com estoque ausente.

**Como pensar:** Zero significa estoque registrado sem unidades; ausência significa quantidade desconhecida.

```python
print("Estoque zero:")
print(df.loc[df["estoque"] == 0, "peca"])
print("Estoque ausente:")
print(df.loc[df["estoque"].isna(), "peca"])
```

**Resultado explicado:** Shorts tem zero unidades. O estoque da Jaqueta é desconhecido.

## Lembretes

| Recurso | Significado |
|---|---|
| `df["coluna"]` | Seleciona uma coluna |
| `df[condicao]` | Seleciona as linhas que atendem à condição |
| `df.loc[linhas, colunas]` | Seleciona por rótulos ou condições |
| `df.iloc[linhas, colunas]` | Seleciona por posições começando em zero |
| `:` sozinho | Todas as linhas ou colunas |
| `0:3` no iloc | Posições 0, 1 e 2; exclui a posição 3 |
| `==` | Compara valores |
| `=` | Atribui um valor |

Selecionar ou mostrar um resultado não significa apagar as outras informações. A alteração de valores é feita explicitamente, como na questão 9.
