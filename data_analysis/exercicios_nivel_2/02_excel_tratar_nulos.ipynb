## Parte 2 — resolução comentada

Compare com suas tentativas. Leia como pensar, execute o código e explique o resultado com suas palavras. Comece executando a preparação novamente.

```python
import pandas as pd

df = pd.read_excel("academia_nulos.xlsx", sheet_name="Alunos")
df
```

### Resolução 1

**Pedido:** Mostre as primeiras linhas e o tamanho da tabela.

**Como pensar:** Lemos a aba indicada e fazemos uma inspeção inicial.

```python
print(df.shape)
df.head()
```

**Resultado explicado:** 6 alunos, 4 colunas; head mostra os cinco primeiros.

### Resolução 2

**Pedido:** Inspecione tipos e valores preenchidos.

**Como pensar:** info resume os tipos e contagens; isna().sum conta ausências.

```python
df.info()
print(df.isna().sum())
```

**Resultado explicado:** plano, mensalidade e visitas têm uma ausência cada.

### Resolução 3

**Pedido:** Mostre nome e mensalidade dos alunos usando loc.

**Como pensar:** Usamos : para todas as linhas e uma lista para duas colunas.

```python
df.loc[:, ["aluno", "mensalidade"]]
```

**Resultado explicado:** A mensalidade de Carla está ausente.

### Resolução 4

**Pedido:** Calcule a média das mensalidades conhecidas.

**Como pensar:** mean ignora a mensalidade ausente.

```python
df["mensalidade"].mean()
```

**Resultado explicado:** A média dos cinco valores conhecidos é 100. Carla não participa do cálculo.

### Resolução 5

**Pedido:** Crie uma cópia e preencha apenas o plano vazio com Não informado.

**Como pensar:** copy preserva a original. fillna substitui ausências apenas na coluna escolhida.

```python
df_tratado = df.copy()
df_tratado["plano"] = df_tratado["plano"].fillna("Não informado")
df_tratado
```

**Resultado explicado:** Bruno passa a ter plano Não informado. As ausências numéricas continuam.

### Resolução 6

**Pedido:** Crie uma tabela só com mensalidades registradas.

**Como pensar:** dropna com subset considera só mensalidade. O resultado é guardado em outra variável.

```python
df_com_mensalidade = df_tratado.dropna(subset=["mensalidade"])
df_com_mensalidade
```

**Resultado explicado:** A tabela resultante tem 5 alunos e exclui Carla. A original tem 6. Diego continua mesmo com visitas ausentes.

### Resolução 7

**Pedido:** Mostre quem precisa de conferência porque não tem mensalidade.

**Como pensar:** Filtramos a ausência para saber quais registros precisam de atualização.

```python
df_tratado.loc[df_tratado["mensalidade"].isna(), ["aluno", "mensalidade"]]
```

**Resultado explicado:** Carla precisa de conferência; ausência não significa que ela não paga.

### Resolução 8

**Pedido:** Compare ausência de visitas com zero visitas.

**Como pensar:** Separar ausência e zero conserva o significado do dado.

```python
print(df_tratado.loc[df_tratado["visitas"].isna(), "aluno"])
print(df_tratado.loc[df_tratado["visitas"] == 0, "aluno"])
```

**Resultado explicado:** Diego tem visitas desconhecidas. Felipe tem zero visitas registradas. Não temos justificativa para substituir a ausência por zero.

### Resolução 9

**Pedido:** Exporte a tabela tratada para CSV.

**Como pensar:** to_csv escreve um arquivo. index=False impede criar uma coluna extra com o índice.

```python
df_tratado.to_csv("academia_tratada.csv", index=False, encoding="utf-8")
```

**Resultado explicado:** Um novo CSV é criado na pasta de trabalho. O Excel de entrada permanece igual. No Colab, clique com o botão direito no CSV gerado e escolha baixar.

## Lembretes

| Recurso | Significado |
|---|---|
| `df["coluna"]` | Seleciona uma coluna |
| `df[condicao]` | Seleciona as linhas que atendem à condição |
| `df.loc[linhas, colunas]` | Seleciona por rótulos ou condições |
| `df.iloc[linhas, colunas]` | Seleciona por posições começando em zero |
| `:` sozinho | Todas as linhas ou colunas |
| `0:3` no iloc | Posições 0, 1 e 2; exclui a posição 3 |
| `==` | Compara valores |
| `=` | Atribui um valor |

Selecionar ou mostrar um resultado não significa apagar as outras informações. A alteração de valores é feita explicitamente, como na questão 9.
