```python
#imports
import pandas as pd
#import: disponibiliza a biblioteca no nosso código.
#pandas: nome da biblioteca.
#pd: apelido que usaremos para escrever menos.
```

```python
#exemplo 1 - lendo de um dicionário 
# Esse dicionário tem três chaves. Cada chave vira uma coluna, 
# e cada lista contém os valores daquela coluna. 

dados = {
    "nome": ["Ana", "Bruno", "Carla", "Diego"],
    "nota": [8, 5, 9, 6],
    "faltas": [2, 5, 1, 3]
}

dados
```

```python
# pd.DataFrame(dados) transforma esses dados em uma tabela. Guardamos a tabela na variável df.

df = pd.DataFrame(dados)

df
```

```python
#visualizar as primeiras linhas
df.head(2)
```

```python
#selecionar uma coluna
df["nota"]
```

```python
#calculo da média
#usando for
soma = 0
quantidade = 0
for nota in df["nota"]:
    soma += nota
    quantidade += 1

media = soma / quantidade
media
```

```python
#outra opção para calcular a média
media_2 = df["nota"].mean()
print(media_2)
```

```python
#alunos com nota acima da média
df_alunos_acima_media = df[df["nota"] > media_2]
df_alunos_acima_media
```

```python
#alunos com nota acima da média
df_alunos_acima_media = df[(df["nota"] > media_2) | (df["faltas"] == 0)]
df_alunos_acima_media
```

```python
#alunos com nota acima da média
df_alunos_acima_media = df[(df["nota"] > media_2) & (df["faltas"] == 0)]
df_alunos_acima_media
```

Comando	-> Como escolhemos os dados
loc	 -> Pelos nomes das colunas, rótulos das linhas ou condições
iloc -> Pela posição das linhas e colunas na tabela

```python
#Na tabela df, selecione as linhas com nota maior ou igual a 7 e mostre somente a coluna nome.”
names = df.loc[df["nota"] > 7, "nome"]
names
```

```python
#Queremos as duas primeiras linhas e as duas primeiras colunas:
df.iloc[:2, :2]
#Antes da vírgula, 0:2: seleciona as linhas nas posições 0 e 1.
#Depois da vírgula, 0:2: seleciona as colunas nas posições 0 e 1.
```

```python
#Selecionar todas as linhas e uma coluna
df.loc[:, "nome"]
# O ':' sozinho significa todas as linhas.
```
