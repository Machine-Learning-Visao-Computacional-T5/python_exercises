## Parte 2 — resolução e como pensar

Agora compare suas tentativas com a resolução. Não basta copiar: identifique a coluna usada, a operação e o significado do resultado.

Execute a preparação novamente para partir da tabela original.

````python
import pandas as pd

dados = {
    "produto": ["Caderno", "Caneta", "Lápis", "Mochila", "Borracha", "Estojo"],
    "preco": [20, 5, 3, 100, 2, 15],
    "estoque": [10, 30, 50, 4, 25, 8]
}

df = pd.DataFrame(dados)
df
````

### 1. Visualizar o começo

**Como pensar:** Queremos visualizar as primeiras linhas. Usamos `head(3)` para pedir três linhas.

````python
df.head(3)
````

**Resultado explicado:** Caderno, Caneta e Lápis aparecem. Três produtos ficam fora da visualização; nenhum é apagado.

### 2. Selecionar uma coluna

**Como pensar:** Precisamos da informação guardada na coluna `produto`. Usamos o nome da coluna entre aspas e colchetes.

````python
df["produto"]
````

**Resultado explicado:** Aparecem os seis nomes. Preços e estoques não aparecem. Selecionar uma coluna desse jeito retorna uma Series.

### 3. Calcular a média

**Como pensar:** Primeiro selecionamos `preco`. Depois usamos `.mean()` para calcular a média e guardamos o número na variável `media`. `print` mostra o resultado.

````python
media = df["preco"].mean()
print(media)
````

**Resultado explicado:** A soma é 145 e a quantidade é 6. A média é aproximadamente R$ 24,17. A média não é o preço de cada produto. O print pode mostrar mais casas decimais.

### 4. Filtrar preços

**Como pensar:** A condição `df["preco"] < 10` verifica cada preço. O `df[...]` externo mantém as linhas nas quais a condição é verdadeira. `<` significa menor que.

Vamos ver a condição separada antes de filtrar: cada `True` significa que aquele produto custa menos de 10; cada `False` significa que ele não atende à condição.

````python
df["preco"] < 10
````

````python
df[df["preco"] < 10]
````

**Resultado explicado:** Caneta, Lápis e Borracha aparecem, com todas as colunas. Um preço igual a 10 ficaria de fora.

### 5. Filtrar e escolher uma coluna com loc

**Como pensar:** No `loc`, antes da vírgula escolhemos as linhas com a condição de estoque. Depois da vírgula escolhemos a coluna `produto`.

````python
df.loc[df["estoque"] < 10, "produto"]
````

**Resultado explicado:** Mochila e Estojo aparecem, com índices originais 3 e 5. Caderno não entra porque seu estoque é exatamente 10.

### 6. Selecionar por posição com iloc

**Como pensar:** No `iloc`, usamos posições começando em 0. Antes da vírgula ficam as linhas: `0:3` inclui 0, 1 e 2. Depois ficam as colunas: `0:2` inclui 0 e 1. O limite final fica de fora.

````python
df.iloc[0:3, 0:2]
````

**Resultado explicado:** Aparecem Caderno, Caneta e Lápis nas colunas produto e preco. A coluna estoque não aparece.

### 7. Selecionar todas as linhas

**Como pensar:** O `:` sozinho significa todas as linhas. Depois da vírgula, `"estoque"` escolhe uma coluna pelo nome.

````python
df.loc[:, "estoque"]
````

**Resultado explicado:** Aparecem os seis valores: 10, 30, 50, 4, 25 e 8, na ordem original.

### 8. Ordenar por preço

**Como pensar:** Usamos `sort_values` para ordenar os valores da coluna `preco`. `ascending=True` pede ordem crescente, do menor para o maior.

````python
df.sort_values("preco", ascending=True)
````

**Resultado explicado:** A ordem é Borracha, Lápis, Caneta, Estojo, Caderno e Mochila. Este código retorna a tabela ordenada; não substitui `df`. Os índices originais são mantidos.

### 9. Alterar uma cópia

**Como pensar:** Criamos uma cópia com `.copy()`. A condição usa `==` para verificar onde o produto é Caderno. Depois da vírgula escolhemos `preco`. O `= 22` atribui o novo valor à seleção. `==` compara; `=` atribui.

````python
df_alterado = df.copy()
df_alterado.loc[df_alterado["produto"] == "Caderno", "preco"] = 22
df_alterado
````

**Resultado explicado:** Na cópia, Caderno custa 22. Os outros valores permanecem iguais. A tabela original continua com preço 20 para Caderno.

### Conferindo que a tabela original foi preservada

Execute:

````python
df.loc[df["produto"] == "Caderno", "preco"]
````

O preço original continua sendo 20. A mudança foi feita somente em `df_alterado`.

## Extra — média com for e com pandas

Com `for`, fazemos a soma e a contagem explicitamente:

````python
soma = 0
quantidade = 0

for preco in df["preco"]:
    soma = soma + preco
    quantidade = quantidade + 1

media_com_for = soma / quantidade
print(media_com_for)
````

O pandas oferece a operação pronta:

````python
media_com_pandas = df["preco"].mean()
print(media_com_pandas)
````

Para estes dados, os resultados são iguais. O `for` mostra os passos do cálculo; `.mean()` expressa o mesmo cálculo com menos código.
