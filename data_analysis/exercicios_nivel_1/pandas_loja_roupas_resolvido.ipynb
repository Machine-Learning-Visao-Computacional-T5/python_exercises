## Parte 2 — resolução comentada

Compare com suas tentativas. Leia como pensar, execute o código e explique o resultado com suas palavras. Comece executando a preparação novamente.

```python
import pandas as pd

dados = {'peca': ['Camiseta', 'Calça', 'Vestido', 'Jaqueta', 'Shorts', 'Saia'], 'preco': [40, 120, 150, 200, 60, 90], 'estoque': [20, 8, 5, 3, 15, 7]}

df = pd.DataFrame(dados)
df
```

### Resolução 1

**Pedido:** Mostre as três primeiras linhas.

**Como pensar:** Queremos ver o começo da tabela; usamos head(3).

```python
df.head(3)
```

**Resultado explicado:** Camiseta, Calça e Vestido aparecem. As demais peças continuam na tabela.

### Resolução 2

**Pedido:** Mostre somente os nomes das peças.

**Como pensar:** Selecionamos uma informação de todas as linhas usando o nome da coluna.

```python
df["peca"]
```

**Resultado explicado:** Aparecem os seis nomes, sem os preços e estoques.

### Resolução 3

**Pedido:** Calcule a média dos preços.

**Como pensar:** Selecionamos preco e aplicamos mean().

```python
media = df["preco"].mean()
print(media)
```

**Resultado explicado:** A soma é 660; 660 / 6 = 110. A média é R$ 110,00, embora nenhuma peça tenha exatamente esse preço.

### Resolução 4

**Pedido:** Mostre as peças que custam menos de R$ 100, com todas as colunas.

**Como pensar:** A condição compara cada preço com 100; os colchetes externos selecionam as linhas verdadeiras.

```python
df[df["preco"] < 100]
```

**Resultado explicado:** Camiseta, Shorts e Saia aparecem. O valor 100 não seria incluído.

### Resolução 5

**Pedido:** Use loc para mostrar somente os nomes das peças com estoque menor ou igual a 5.

**Como pensar:** Antes da vírgula usamos a condição; depois escolhemos peca. <= inclui o limite.

```python
df.loc[df["estoque"] <= 5, "peca"]
```

**Resultado explicado:** Vestido e Jaqueta aparecem. Vestido entra porque tem estoque 5.

### Resolução 6

**Pedido:** Use iloc para mostrar as duas primeiras linhas e as duas primeiras colunas.

**Como pensar:** As posições começam em 0. O final 2 fica de fora.

```python
df.iloc[0:2, 0:2]
```

**Resultado explicado:** Camiseta e Calça aparecem nas colunas peca e preco.

### Resolução 7

**Pedido:** Mostre os estoques de todas as peças usando loc.

**Como pensar:** O : escolhe todas as linhas; estoque escolhe a coluna.

```python
df.loc[:, "estoque"]
```

**Resultado explicado:** Os valores são 20, 8, 5, 3, 15 e 7.

### Resolução 8

**Pedido:** Ordene do maior preço para o menor.

**Como pensar:** sort_values ordena a coluna. ascending=False pede ordem decrescente.

```python
df.sort_values("preco", ascending=False)
```

**Resultado explicado:** Jaqueta, Vestido, Calça, Saia, Shorts e Camiseta. O resultado não substitui df.

### Resolução 9

**Pedido:** Crie uma cópia e altere o preço da Camiseta para R$ 45.

**Como pensar:** copy cria a cópia. == identifica a peça; = atribui o novo preço à coluna selecionada.

```python
df_alterado = df.copy()
df_alterado.loc[df_alterado["peca"] == "Camiseta", "preco"] = 45
df_alterado
```

**Resultado explicado:** A cópia tem Camiseta a 45; a original continua com 40.

### Resolução 10

**Pedido:** Desafio: calcule o valor do estoque de cada peça em uma nova coluna de uma cópia.

**Como pensar:** Multiplicamos preço por quantidade, linha a linha, e guardamos em uma nova coluna.

```python
df_valores = df.copy()
df_valores["valor_estoque"] = df_valores["preco"] * df_valores["estoque"]
df_valores
```

**Resultado explicado:** Valores: 800, 960, 750, 600, 900 e 630 reais. Isso é valor de estoque ao preço de venda, não lucro.

## Lembretes

| Recurso | Significado |
|---|---|
| `df["coluna"]` | Seleciona uma coluna |
| `df[condicao]` | Seleciona as linhas que atendem à condição |
| `df.loc[linhas, colunas]` | Seleciona por rótulos ou condições |
| `df.iloc[linhas, colunas]` | Seleciona por posições começando em zero |
| `:` sozinho | Todas as linhas ou colunas |
| `0:3` no iloc | Posições 0, 1 e 2; exclui a posição 3 |
| `==` | Compara valores |
| `=` | Atribui um valor |

Selecionar ou mostrar um resultado não significa apagar as outras informações. A alteração de valores é feita explicitamente, como na questão 9.
