## 1. O que é NumPy?

**NumPy é uma biblioteca do Python para trabalhar com números e arrays.** Uma biblioteca reúne ferramentas prontas que podemos usar no nosso programa.

Um **array** é uma estrutura que guarda vários valores. Neste início, pense nele como uma sequência de números na qual podemos aplicar cálculos a todos os elementos de uma vez. Também podemos criar arrays com linhas e colunas.

Em um array comum do NumPy, os elementos compartilham um tipo de dado (`dtype`). Uma lista Python pode misturar tipos livremente; um array numérico é organizado para cálculos.

`import numpy as np` importa a biblioteca e dá a ela o apelido `np`. Esse apelido é uma convenção: não cria outra biblioteca.

```python
import numpy as np

print("NumPy importado!")
```

Se apareceu a mensagem, podemos usar os comandos que começam com `np`.

### Para que NumPy é bom?

NumPy ajuda quando precisamos **fazer cálculos com muitos números seguindo a mesma regra**. Podemos escrever uma operação para o conjunto inteiro, em vez de escrever um cálculo separado para cada valor.

| Situação | Como NumPy ajuda |
|---|---|
| Preços de uma loja | Aplicar desconto ou aumento a todos os preços |
| Compras | Multiplicar preços por quantidades e somar o total |
| Notas de uma turma | Calcular média, mínimo, máximo e selecionar notas |
| Medições | Calcular e comparar conjuntos de temperaturas ou distâncias |
| Imagens | Representar os valores dos pixels em arrays e fazer cálculos sobre eles |
| Machine learning | Representar características numéricas e trabalhar com vetores e matrizes |

**Exemplo concreto:** uma loja quer dar 10% de desconto em três preços. Com 10% de desconto, pagamos 90% do preço: multiplicamos por `0.90`.


```python
precos_desconto = np.array([100, 200, 50])
precos_com_desconto = precos_desconto * 0.90

print("Antes:", precos_desconto)
print("Com 10% de desconto:", precos_com_desconto)
```

O resultado é `[90. 180. 45.]`. Uma operação aplica a mesma regra aos três preços. O array original continua com 100, 200 e 50.

**Por que usar?** O código expressa o cálculo de forma curta e as operações numéricas do NumPy costumam ser eficientes para grandes arrays. Isso não significa que qualquer programa ficará mais rápido só por importar a biblioteca.

**Quando uma lista já resolve?** Para guardar nomes, organizar poucos itens ou acrescentar e remover elementos com frequência, uma lista Python pode ser suficiente.

**E o pandas?** Pandas é conveniente para tabelas com nomes de colunas, como nome, idade e cidade. NumPy se concentra em arrays e cálculos numéricos. As duas bibliotecas podem ser usadas juntas; não precisamos escolher uma para todas as tarefas.

## 3. Lista Python × array NumPy

O símbolo `*` pode ter efeitos diferentes conforme o tipo do objeto. Em uma lista, multiplicar por 2 repete a sequência. Em um array numérico, multiplica cada elemento por 2.

### O que uma lista e um array têm em comum?

Você já conhece listas? Então já conhece várias ideias que também usaremos com arrays! **Os dois podem guardar vários valores em uma sequência ordenada.** Ordenada significa que cada elemento tem uma posição; não significa que os valores estejam do menor para o maior.

Vamos representar os mesmos três preços de duas maneiras:

| Semelhança | Lista Python | Array NumPy |
|---|---|---|
| Guarda vários valores | `[10, 20, 30]` | `np.array([10, 20, 30])` |
| Primeiro índice é zero | `lista_precos[0]` | `array_precos[0]` |
| Pode acessar o último com -1 | `lista_precos[-1]` | `array_precos[-1]` |
| Permite selecionar trechos | `lista_precos[0:2]` | `array_precos[0:2]` |
| Pode consultar o tamanho | `len(lista_precos)` | `len(array_precos)` |
| Permite percorrer com for | `for preco in lista_precos:` | `for preco in array_precos:` |
| Permite alterar um elemento | `lista_precos[0] = 15` | `array_precos[0] = 15` |

Aqui usamos um array de **uma dimensão**. Nesse caso, `len` informa a quantidade de elementos. Em um array de duas dimensões, `len` informa o tamanho da primeira dimensão, enquanto `.size` informa o total de elementos.


```python
lista_precos = [10, 20, 30]
array_precos = np.array(lista_precos)

print("Lista:", lista_precos)
print("Array:", array_precos)
print("Primeiro da lista:", lista_precos[0])
print("Primeiro do array:", array_precos[0])
print("Último da lista:", lista_precos[-1])
print("Último do array:", array_precos[-1])
print("Trecho da lista:", lista_precos[0:2])
print("Trecho do array:", array_precos[0:2])
print("Tamanho da lista:", len(lista_precos))
print("Tamanho do array:", len(array_precos))
```

**Leia o código:** `lista_precos[0]` e `array_precos[0]` usam a mesma ideia: acessar o valor da primeira posição. Nos dois, `[0:2]` seleciona os índices 0 e 1; o índice 2 fica de fora.

Os resultados têm os mesmos valores, mas os objetos são de tipos diferentes. As vírgulas aparecem na impressão da lista; normalmente não aparecem na impressão do array.

Também podemos usar `for` para visitar um valor por vez nos dois objetos.

```python
print("Percorrendo a lista:")
for preco in lista_precos:
    print(preco)

print("Percorrendo o array:")
for preco in array_precos:
    print(preco)
```

`preco` recebe um elemento por vez: primeiro 10, depois 20 e depois 30. O bloco com recuo é executado uma vez para cada elemento.

Os dois também permitem alterar um valor por posição:

```python
lista_precos[0] = 15
array_precos[0] = 15

print("Lista alterada:", lista_precos)
print("Array alterado:", array_precos)
```

Agora os dois contêm 15, 20 e 30. Estamos alterando um elemento que já existe, não acrescentando um novo.

### Semelhanças não significam comportamento idêntico

- **Cálculos:** lista multiplicada por 2 repete os elementos; array numérico multiplicado por 2 dobra cada valor. Vamos conferir isso no exemplo seguinte.
- **Tipos:** uma lista pode misturar tipos; os elementos de um array comum compartilham um `dtype`. Ao criar um array com tipos diferentes, o NumPy pode converter os valores para um tipo comum.
- **Acrescentar:** listas têm `.append()`. Arrays NumPy não têm esse método; `np.append` cria um novo array com os valores acrescentados.
- **Fatiamento:** um trecho de lista cria uma nova lista superficial. Um fatiamento simples de array geralmente compartilha os dados com o original. Para editar o trecho de array de forma independente, use `.copy()`, explicado mais adiante.

**Pergunta para conferir:** se os valores são `[10, 20, 30]`, o que `[1]` seleciona em cada estrutura? **20 nos dois casos**, pois a contagem começa em zero.


```python
lista = [10, 20, 30]
array = np.array(lista)

print("Lista vezes 2:", lista * 2)
print("Array vezes 2:", array * 2)
```

A lista resulta em `[10, 20, 30, 10, 20, 30]`. O array resulta em `[20 40 60]`. Por isso precisamos observar o tipo do dado antes de prever o cálculo.

## 4. Atributo, função e método

**Atributo** é uma informação do objeto: usamos sem parênteses, como `precos.size`.

**Função** é uma operação que chamamos com parênteses, como `np.sum(precos)`.

**Método** é uma operação acessada pelo próprio objeto, como `precos.sum()`.

**Parâmetro** é o nome de uma entrada na definição de uma função; **argumento** é o valor passado na chamada. **Retorno** é o resultado que a chamada entrega para ser usado ou guardado.

| Expressão | O que informa |
|---|---|
| `precos.ndim` | Número de dimensões |
| `precos.shape` | Tamanho de cada dimensão |
| `precos.size` | Total de elementos |
| `precos.dtype` | Tipo dos elementos |

```python
print("Dimensões:", precos.ndim)
print("Formato:", precos.shape)
print("Elementos:", precos.size)
print("Tipo dos elementos:", precos.dtype)

print("Soma com função:", np.sum(precos))
print("Soma com método:", precos.sum())
```

Este array tem uma dimensão, formato `(3,)` e três elementos. `(3,)` é uma tupla com um único tamanho: 3. A vírgula faz parte da escrita de uma tupla de um elemento.

O tipo inteiro exibido pode variar conforme o ambiente. As duas formas de somar devolvem 60.

`precos.sum` sem `()` acessa o método, mas não executa a soma. `precos.size()` tentaria chamar um número e causaria erro: `size` é atributo.

## 5. Acessar elementos com índices

Um **índice** indica a posição de um elemento. Em Python, a primeira posição é **0**. Os colchetes agora acessam elementos; na criação do array, eles formavam uma lista.

```python
idades = np.array([18, 25, 32, 40])

print("Primeiro:", idades[0])
print("Segundo:", idades[1])
print("Último:", idades[-1])
```

`idades[0]` devolve 18; `idades[1]` devolve 25; `idades[-1]` devolve 40. O índice negativo -1 conta a partir do final. `idades[4]` causaria erro, pois os índices válidos positivos são 0, 1, 2 e 3.

## 6. Selecionar um trecho: fatiamento

O fatiamento seleciona um trecho: `array[início:fim]`. **O início entra; o fim não entra.**

Em `idades[1:3]`, selecionamos os índices 1 e 2. Se omitirmos o início, começamos no primeiro elemento; se omitirmos o fim, vamos até o final.

```python
print("Índices 1 e 2:", idades[1:3])
print("Do começo até antes do índice 2:", idades[:2])
print("Do índice 2 até o final:", idades[2:])
```

Os resultados são `[25 32]`, `[18 25]` e `[32 40]`, respectivamente.

## 7. Calcular com todos os elementos

Podemos somar, subtrair, multiplicar ou dividir cada elemento por um número. Esse cálculo por elemento costuma evitar um `for` escrito por nós.

A operação abaixo produz um novo resultado; guardar em outra variável mantém o array original neste exemplo.

```python
valores = np.array([10, 20, 30])

print("Mais 5:", valores + 5)
print("Menos 2:", valores - 2)
print("Dobro:", valores * 2)
print("Metade:", valores / 2)

valores_com_aumento = valores + 5
print("Original:", valores)
print("Com aumento:", valores_com_aumento)
```

Os resultados são `[15 25 35]`, `[8 18 28]`, `[20 40 60]` e `[5. 10. 15.]`. A divisão `/` produz números com parte decimal. O ponto em `5.` indica 5.0.

## 8. Operações entre dois arrays

Para este exemplo, cada preço tem uma quantidade correspondente. Como os dois arrays têm o mesmo formato, a multiplicação combina os elementos pela posição.

```python
precos_compra = np.array([10, 20, 5])
quantidades = np.array([2, 3, 4])

subtotais = precos_compra * quantidades
print("Subtotais:", subtotais)
print("Total da compra:", subtotais.sum())
```

O cálculo é 10 × 2, 20 × 3 e 5 × 4: `[20 60 20]`. A soma é 100. Não é multiplicação de matrizes. Outros formatos exigem regras de compatibilidade; aqui usamos arrays iguais para manter o exemplo simples.

## 9. Somar, calcular média, mínimo e máximo

Estas funções resumem um conjunto de números.

| Função | Uso |
|---|---|
| `np.sum` | Soma |
| `np.mean` | Média aritmética |
| `np.min` | Menor valor |
| `np.max` | Maior valor |

A média é a soma dividida pela quantidade de valores.

```python
notas = np.array([6, 8, 10])

print("Soma:", np.sum(notas))
print("Média:", np.mean(notas))
print("Menor:", np.min(notas))
print("Maior:", np.max(notas))
```

A soma é 24; a média é 24 ÷ 3 = 8.0; o mínimo é 6 e o máximo é 10.

## 10. Comparações e filtros

`>=` significa maior ou igual. A comparação abaixo devolve um array de valores **booleanos**: `True` significa verdadeiro; `False`, falso.

Essa sequência é uma **máscara**: ao colocá-la entre colchetes, selecionamos apenas os elementos marcados com `True`.

```python
idades_filtro = np.array([18, 25, 32, 40])
mascara = idades_filtro >= 25

print("Condição:", mascara)
selecionadas = idades_filtro[mascara]
print("Selecionadas:", selecionadas)

# A mesma seleção em uma linha:
print(idades_filtro[idades_filtro >= 25])
```

A máscara é `[False True True True]`. O filtro devolve `[25 32 40]`. A idade 25 entra porque usamos `>=`, e não apenas `>`.

## 11. Escolher um resultado com np.where

`np.where(condição, valor_se_verdadeiro, valor_se_falso)` escolhe um resultado para cada posição.

Vamos aplicar a regra: nota maior ou igual a 7 recebe `"Aprovado"`; caso contrário, recebe `"Revisar"`. As aspas indicam texto.

```python
notas_status = np.array([6, 7, 9])
status = np.where(notas_status >= 7, "Aprovado", "Revisar")

print(status)
```

O resultado é `['Revisar' 'Aprovado' 'Aprovado']`. O 7 atende à condição porque o limite está incluído. `np.where` devolve um novo array; não altera as notas.

## 12. Criar sequências e ordenar

`np.arange(início, fim, passo)` cria uma sequência. O início entra e o fim não entra. O passo informa quanto avançamos.

`np.sort(array)` devolve os valores em ordem crescente.

```python
sequencia = np.arange(0, 10, 2)
print("Sequência:", sequencia)

valores_fora_de_ordem = np.array([30, 10, 20])
ordenados = np.sort(valores_fora_de_ordem)
print("Ordenados:", ordenados)
print("Original:", valores_fora_de_ordem)
```

A sequência é `[0 2 4 6 8]`: o 10 não entra. A ordenação devolve `[10 20 30]`. `np.sort` mantém o array original neste exemplo.

## 13. Arrays com linhas e colunas

Um array de duas dimensões pode ser visto como uma grade. Cada lista interna representa uma linha.

`matriz[linha, coluna]` seleciona uma posição; ambas as contagens começam em zero. `:` significa selecionar todos os elementos daquela dimensão.

```python
matriz = np.array([[10, 20], [30, 40]])
print(matriz)
print("Formato:", matriz.shape)
print("Dimensões:", matriz.ndim)
print("Elementos:", matriz.size)
print("Linha 0, coluna 1:", matriz[0, 1])
print("Primeira linha:", matriz[0, :])
print("Primeira coluna:", matriz[:, 0])
```

A matriz tem formato `(2, 2)`, duas dimensões e quatro elementos. A posição `[0, 1]` contém 20. A primeira linha é `[10 20]` e a primeira coluna é `[10 30]`.

## 14. Cuidado ao copiar um array

`copia = original` cria outro nome para o mesmo objeto. Para ter um array independente neste exemplo, use `original.copy()`.

Fatiamentos simples de arrays NumPy geralmente também compartilham dados com o original. Se você pretende editar o trecho sem alterar o original, use `.copy()`.

```python
original = np.array([10, 20, 30])
copia_independente = original.copy()
copia_independente[0] = 99

print("Original:", original)
print("Cópia:", copia_independente)
```

A cópia fica `[99 20 30]`; o original continua `[10 20 30]`. A atribuição com `[0] = 99` altera o elemento da primeira posição na cópia.

## 19. Resumão de NumPy

**NumPy é uma biblioteca do Python para cálculos numéricos.** Seu objeto principal é o **array**, que guarda valores em uma ou mais dimensões. Importamos com `import numpy as np`; `np` é o apelido usado para acessar a biblioteca.

### Ideias para lembrar

- **Lista e array:** ambos guardam sequências, permitem índices, fatiamento e `for`. Em uma lista, `* 2` repete a sequência; em um array numérico, dobra cada elemento.
- **Índices:** começam em zero. `[0]` acessa o primeiro elemento; `[-1]`, o último.
- **Fatiamento:** `[início:fim]` inclui o início e exclui o fim.
- **Cálculos:** `array + 5` e `array * 2` aplicam a operação a cada elemento.
- **Comparação:** `array >= 10` devolve uma máscara de `True` e `False`.
- **Filtro:** `array[array >= 10]` seleciona os valores que atendem à condição.
- **Formato:** `shape` descreve o tamanho de cada dimensão; `ndim` conta dimensões; `size` conta todos os elementos; `dtype` informa o tipo dos elementos.
- **Cópia:** use `.copy()` para obter um array independente. Uma atribuição simples não copia o objeto; fatiamentos simples de arrays geralmente compartilham dados.

### Comandos com exemplos e resultados

Considere `a = np.array([10, 20, 30])`.

| Objetivo | Comando | Resultado |
|---|---|---|
| Criar um array | `np.array([10, 20, 30])` | Array com três valores |
| Primeiro valor | `a[0]` | `10` |
| Último valor | `a[-1]` | `30` |
| Dois primeiros | `a[:2]` | `[10 20]` |
| Quantidade de elementos | `a.size` | `3` |
| Formato | `a.shape` | `(3,)` |
| Dimensões | `a.ndim` | `1` |
| Somar 5 a cada valor | `a + 5` | `[15 25 35]` |
| Dobrar os valores | `a * 2` | `[20 40 60]` |
| Somar todos | `np.sum(a)` ou `a.sum()` | `60` |
| Calcular média | `np.mean(a)` | `20.0` |
| Menor valor | `np.min(a)` | `10` |
| Maior valor | `np.max(a)` | `30` |
| Comparar | `a >= 20` | `[False True True]` |
| Filtrar | `a[a >= 20]` | `[20 30]` |

### Como ler a escrita

| Símbolo ou termo | Significado | Exemplo |
|---|---|---|
| `.` | Acessa algo da biblioteca ou do objeto | `np.array`, `a.shape` |
| `()` | Executa uma função ou método | `a.sum()` |
| `[]` | Cria uma lista ou seleciona valores, dependendo do contexto | `[10, 20]`, `a[0]` |
| `:` | Delimita um trecho em um fatiamento | `a[0:2]` |
| `=` | Guarda um valor em uma variável | `total = a.sum()` |
| `==` | Compara igualdade | `a == 20` |
| Função | Operação chamada com argumentos | `np.mean(a)` |
| Método | Operação acessada pelo objeto | `a.sum()` |
| Atributo | Informação consultada sem parênteses | `a.size` |
| Argumento | Valor passado na chamada | `a` em `np.mean(a)` |
| Retorno | Resultado entregue pela chamada | `20.0` em `np.mean(a)` |

### Uma dimensão e duas dimensões

- `np.array([10, 20, 30])`: uma dimensão, formato `(3,)`.
- `np.array([[10, 20], [30, 40]])`: duas dimensões, formato `(2, 2)`.
- Em uma matriz, `[linha, coluna]` seleciona uma posição. A contagem começa em zero para as duas.
- `len(array)` informa o tamanho da primeira dimensão; `array.size` informa o total de elementos.

### Quando usar?

Use NumPy para conjuntos de números, cálculos por elemento, estatísticas e matrizes. Listas são úteis para sequências gerais; pandas é conveniente para tabelas com nomes de colunas. Essas ferramentas podem trabalhar juntas.
