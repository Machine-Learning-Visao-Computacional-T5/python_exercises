## Qual comando escolher?
| Comando | Quando usar | Como ler |
|---|---|---|
| `if` | Fazer algo se uma condição for verdadeira | Se… |
| `else` | Definir o outro caminho quando a condição anterior for falsa | Caso contrário… |
| `elif` | Testar outra condição quando as anteriores forem falsas | Senão, se… |
| `for` | Percorrer os itens de uma lista ou outra sequência | Para cada item… |
| `range` | Produzir uma sequência de inteiros, útil com `for` | De um início até antes do fim… |
| `while` | Repetir enquanto uma condição continuar verdadeira | Enquanto… |

`range` não repete o código sozinho: é o `for` que faz a repetição nos exemplos.

## Pequenos detalhes que vamos usar
- `=` guarda um valor: `nota = 6`.
- `==` compara: `nota == 6` pergunta se a nota é igual a 6.
- `>` maior; `<` menor; `>=` maior ou igual; `<=` menor ou igual; `!=` diferente.
- Condições produzem `True` (verdadeiro) ou `False` (falso).
- O `:` inicia um bloco. Use **4 espaços** nas linhas dentro dele.
- `print` mostra um resultado. Textos ficam entre aspas; números não precisam delas.

## 5. `while` — repetir enquanto a condição for verdadeira

**Quando usar:** quando a repetição depende de uma condição; é preciso acompanhar o que muda e como ela termina.

**Exemplo guiado:** Mostrar uma contagem de 1 a 3.

Antes de executar, tente prever a mensagem ou os valores.

````python
contador = 1
while contador <= 3:
    print(contador)
    contador = contador + 1
print("Fim")
````

**Leia o código:** Antes de cada volta, Python verifica a condição. Mostra 1 e muda o contador para 2; mostra 2 e muda para 3; mostra 3 e muda para 4. `4 <= 3` é falso e o laço termina. A atualização é necessária aqui para chegar à parada. Se o contador começar em 4, nenhuma volta acontece. Não execute esta contagem sem a linha que atualiza o contador; se um laço não terminar, use o botão de interromper execução do notebook.

### Acompanhe o `while` à mão
| Contador antes do teste | `contador <= 3` | O que acontece |
|---|---|---|
| 1 | Verdadeiro | Mostra 1; contador vira 2 |
| 2 | Verdadeiro | Mostra 2; contador vira 3 |
| 3 | Verdadeiro | Mostra 3; contador vira 4 |
| 4 | Falso | Sai do laço; mostra Fim |

Na contagem, tanto `for` quanto `while` poderiam funcionar. Com `for` e `range`, os valores vêm da sequência; com `while`, nós atualizamos o contador.

## Resolução 10 — Juntar dinheiro

**Plano:** O saldo muda a cada volta. Ao chegar a 30, saldo < meta é falso. A condição é verificada antes de cada repetição.

**Saída esperada:**
```text
Saldo: 10
Saldo: 20
Saldo: 30
Meta atingida
```

````python
saldo = 0
meta = 30
while saldo < meta:
    saldo = saldo + 10
    print("Saldo:", saldo)
print("Meta atingida")
````

**Verificação e nova tentativa:** Use meta 25. Por que o saldo termina em 30? Depois comece com saldo 30 e meta 30.

## Resolução 11 — Classificar todos os preços

**Plano:** O for traz um preço por vez. Dentro de cada volta, o if escolhe a mensagem. Os prints usam 8 espaços porque estão dentro de dois blocos.

**Saída esperada:**
```text
5 Até 10
10 Até 10
15 Acima de 10
```

````python
precos = [5, 10, 15]
for preco in precos:
    if preco <= 10:
        print(preco, "Até 10")
    else:
        print(preco, "Acima de 10")
````

**Verificação e nova tentativa:** Use [9, 10, 11]. Depois altere a regra para “Menor que 10”, ajustando também as mensagens.

## Resolução 12 — Contar aprovados

**Plano:** A nota 8 aumenta a contagem para 1; 5 não altera; 6 aumenta para 2; 3 não altera. Estou contando pessoas, não somando as notas.

**Saída esperada:**
```text
Quantidade de aprovados: 2
```

````python
notas = [8, 5, 6, 3]
quantidade = 0
for nota in notas:
    if nota >= 6:
        quantidade = quantidade + 1
print("Quantidade de aprovados:", quantidade)
````

**Verificação e nova tentativa:** Use [6, 6, 5]. Depois [2, 4]. Explique a diferença entre somar valores e contar itens.

## Erros comuns e como voltar ao fluxo
| Situação | O que conferir |
|---|---|
| `SyntaxError` | Dois-pontos, aspas, parênteses e operador de comparação |
| `IndentationError` | Os espaços no início das linhas; use 4 por nível |
| `NameError` | Se criou a variável e escreveu exatamente o mesmo nome |
| A nota 6 foi excluída indevidamente | Se usou `>` quando a regra pede `>=` |
| A mensagem final aparece várias vezes | Se o print final ficou dentro do laço |
| O total está sempre errado | Se reiniciou o acumulador dentro da repetição |
| O while continua executando | Interrompa; confira a atualização e a condição de parada |
