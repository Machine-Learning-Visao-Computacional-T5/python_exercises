## Qual estrutura escolher?
| Estrutura | Como escrever | Quando usar | Como acessar |
|---|---|---|---|
| Lista | `[10, 20, 30]` | Sequência que pode receber alterações e repetidos | Posição: `precos[0]` |

Uma lista e uma tupla começam na posição **0**. O segundo item fica em 1. Um conjunto não tem posição; não use `temas[0]`. Para um conjunto vazio, use `set()`: `{}` é um dicionário vazio.

JSON é um formato de texto. Um dicionário Python e um texto JSON podem parecer parecidos, mas são tipos diferentes.

## JSON — exemplo explicado

`import json` disponibiliza o módulo. `json.loads(texto)` transforma texto JSON em dados Python. `json.dumps(dados)` faz o caminho inverso. O ponto aqui acessa uma função dentro do módulo `json`.

| JSON | Python |
|---|---|
| objeto `{...}` | dicionário |
| array `[...]` | lista |
| string | texto (`str`) |
| número inteiro / decimal | `int` / `float` |
| `true` / `false` | `True` / `False` |
| `null` | `None` |

No texto JSON, chaves e strings usam aspas duplas. Não há vírgula depois do último item. As aspas simples externas nos exemplos são delimitadores da string Python, não fazem parte do conteúdo JSON. `loads` e `dumps` trabalham com texto; `load` e `dump` trabalham com arquivos abertos. Nesta aula usaremos texto.

Preveja a saída antes de executar.

````python
import json
texto = '{"nome": "Ana", "ativo": true, "telefone": null}'
aluno = json.loads(texto)
print(type(texto).__name__)
print(type(aluno).__name__)
print(aluno["nome"])
print(aluno["ativo"])
print(aluno["telefone"])
````

## Resolução 15 — Filtrar produtos do JSON

**Plano e explicação:** O JSON começa com um array, então loads devolve uma lista. Percorro os dicionários e guardo apenas os nomes que passam na regra. O preço 10 entra.

**Saída esperada:**
```text
['Caneta', 'Lápis']
```

````python
import json
texto = '[{"nome": "Caneta", "preco": 5}, {"nome": "Caderno", "preco": 20}, {"nome": "Lápis", "preco": 10}]'
produtos = json.loads(texto)
nomes = []
for produto in produtos:
    if produto["preco"] <= 10:
        nomes.append(produto["nome"])
print(nomes)
````

**Nova tentativa:** Mude a regra para preço < 10. Depois teste o texto JSON [].

## Resolução 16 — Total de um pedido

**Plano e explicação:** Converto, acesso a lista de itens e somo um preço por vez. Neste exercício cada item representa uma unidade; não há campo de quantidade.

**Saída esperada:**
```text
Total: 25
```

````python
import json
texto = '{"pedido": 101, "itens": [{"nome": "Caneta", "preco": 5}, {"nome": "Caderno", "preco": 20}]}'
pedido = json.loads(texto)
itens = pedido["itens"]
total = 0
for item in itens:
    total = total + item["preco"]
print("Total:", total)
````

**Nova tentativa:** Acrescente um produto de preço 3. Depois use itens vazio e preveja o total.

## Se algo der errado
| Mensagem ou problema | O que verificar |
|---|---|
| `IndexError` | A posição existe? O índice começa em 0? |
| `KeyError` | A chave está no dicionário? O nome coincide? |
| `TypeError` ao acessar uma string por chave | Você converteu o texto JSON antes de consultar? |
| `JSONDecodeError` | Aspas duplas internas, vírgulas, chaves, colchetes e true/false/null |
| Lista virou None depois de append | append altera a lista, não devolve a lista atualizada |
| Conjunto apareceu em outra ordem | Não dependa da ordem; use sorted para apresentação |
| Resultado antigo | Execute novamente as células após mudar os dados |

O padrão de get só é usado quando a chave está ausente. Uma chave existente com None não é substituída automaticamente.
