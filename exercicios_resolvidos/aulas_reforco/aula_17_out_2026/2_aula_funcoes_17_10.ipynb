## 1. O que é uma função?

Uma função é um **pedaço de código com nome**, que você escreve uma vez e usa quantas vezes quiser.

Pense numa **máquina de suco**:

- Você coloca a fruta (a **entrada**, chamada de *parâmetro*).
- A máquina faz o trabalho lá dentro (o **corpo** da função).
- Sai o suco (a **saída**, chamada de *retorno*).

Você não precisa saber como a máquina funciona por dentro toda vez. Só aperta o botão.

**Por que usar funções?**

- **Não repetir código.** Escreveu uma vez, usa mil vezes.
- **Organizar.** Cada função faz uma coisa, com um nome que explica o quê.
- **Consertar num lugar só.** Achou um erro? Corrige na função e todo lugar que usa ela fica certo.

## 3. Criando sua primeira função

```python
def dizer_oi():
    print("Oi! Bem-vindo à aula de sábado.")
```

Peça por peça:

| Pedaço | O que significa |
| --- | --- |
| `def` | Abreviação de *define*. Avisa o Python: "vou criar uma função" |
| `dizer_oi` | O nome que você escolheu. Use letras minúsculas e `_` no lugar de espaço |
| `()` | Os parênteses onde ficam as entradas. Aqui está vazio: essa função não precisa de nada |
| `:` | Os dois-pontos dizem "agora vem o corpo da função". **Obrigatório** |
| 4 espaços no começo | A **indentação**. Tudo que está recuado faz parte da função. É assim que o Python sabe onde ela começa e termina |

Rode a célula abaixo. **Não vai aparecer nada**, e está certo: você só *criou* a máquina, ainda não ligou.

````python
def dizer_oi():
    print("Oi! Bem-vindo à aula de sábado.")
````

## 5. Parâmetros: entregando coisas para a função

Uma função fica muito mais útil quando recebe informação. O que vai entre os parênteses na hora de criar se chama **parâmetro**. É como uma caixinha vazia com nome, que recebe um valor quando a função é chamada.

````python
def cumprimentar(nome):
    print("Oi, " + nome + "! Bom sábado!")

cumprimentar("Ana")
cumprimentar("Bruno")
cumprimentar("Carla")
````

O que aconteceu:

1. Na chamada `cumprimentar("Ana")`, o valor `"Ana"` foi colocado na caixinha `nome`.
2. Dentro da função, `nome` vale `"Ana"`.
3. Na próxima chamada, a caixinha recebe `"Bruno"`, e assim por diante.

**Vocabulário que cai em prova:** `nome` (na criação) é o **parâmetro**. `"Ana"` (na chamada) é o **argumento**.

## 6. Vários parâmetros

Separe com vírgula. Na chamada, os valores entram **na mesma ordem**.

````python
def apresentar(nome, idade, cidade):
    print(nome, "tem", idade, "anos e mora em", cidade)

apresentar("Ana", 25, "Curitiba")
apresentar("Bruno", 31, "Londrina")
````

A ordem importa! Troque e veja o resultado estranho:

````python
apresentar(25, "Curitiba", "Ana")   # ordem errada: o Python não adivinha
````

## 7. `return`: a função devolvendo um resultado

Até agora as funções só **mostravam** coisas com `print`. Mas muitas vezes você quer que a função **calcule e devolva** um valor, para você guardar numa variável e usar depois. Para isso existe o `return`.

````python
def somar(a, b):
    return a + b

resultado = somar(2, 3)
print("O resultado é", resultado)
print("O dobro do resultado é", resultado * 2)
````

O `return` é a saída da máquina de suco: o valor **sai** da função e você pode guardar, somar, comparar, passar para outra função.

Você pode usar o retorno direto, sem variável:

````python
print(somar(10, 20))
print(somar(somar(1, 2), somar(3, 4)))   # o retorno de uma vira entrada da outra
````

## 8. A confusão número 1: `print` × `return`

Isso derruba todo iniciante. Compare as duas funções:

````python
def dobro_com_print(x):
    print(x * 2)

def dobro_com_return(x):
    return x * 2

a = dobro_com_print(5)    # mostra 10 na tela...
b = dobro_com_return(5)   # não mostra nada...

print("a vale:", a)
print("b vale:", b)
````

Surpresa: **`a` vale `None`** (nada)! A função com `print` só *mostrou* o 10 na tela, mas não *devolveu* nada. Já a função com `return` devolveu o 10, que ficou guardado em `b`.

| | `print` | `return` |
| --- | --- | --- |
| O que faz | Mostra na tela para **você** ver | Devolve o valor para o **programa** usar |
| Dá para guardar numa variável? | Não (a variável fica `None`) | Sim |
| Dá para fazer conta com o resultado? | Não | Sim |
| Encerra a função? | Não | **Sim.** O que vem depois do `return` não roda |

**Regra prática:** se o resultado vai ser **usado** depois, use `return`. Use `print` só para mostrar mensagens.

````python
def teste():
    return "cheguei no return"
    print("esta linha NUNCA roda")   # depois do return, a função já acabou

print(teste())
````

## 9. Valor padrão

Dá para deixar um parâmetro com um valor **padrão**. Se a pessoa não informar, usa o padrão.

````python
def cumprimentar(nome, saudacao="Bom dia"):
    return saudacao + ", " + nome + "!"

print(cumprimentar("Ana"))                   # usa o padrão
print(cumprimentar("Ana", "Boa noite"))      # troca o padrão
````

⚠️ Parâmetros com valor padrão vêm **sempre por último** na lista.

## 11. Variáveis de dentro ficam dentro

Uma variável criada **dentro** de uma função só existe lá dentro. Quando a função termina, ela some. Isso se chama **escopo**.

````python
def calcular_media(n1, n2):
    media_calculada = (n1 + n2) / 2
    return media_calculada

print(calcular_media(7, 9))

try:
    print(media_calculada)   # tentando usar a variável fora da função
except NameError as erro:
    print("Erro:", erro)
````

A variável `media_calculada` existiu só durante a função. Fora dela o Python não conhece. Para usar o valor fora, é para isso que serve o `return`: guarde o resultado numa variável **fora** da função.

## 12. Documentando a função (docstring)

Logo abaixo do `def`, escreva entre três aspas o que a função faz. É uma boa prática, e o Python mostra esse texto quando você pede ajuda.

````python
def area_retangulo(base, altura):
    """Calcula a área de um retângulo: base vezes altura."""
    return base * altura

print(area_retangulo(3, 4))
help(area_retangulo)
````

## 13. Deu ruim: erros comuns

| Mensagem ou sintoma | O que aconteceu | Como resolver |
| --- | --- | --- |
| `SyntaxError: expected ':'` | Esqueceu os dois-pontos no fim do `def` | `def somar(a, b):` |
| `IndentationError: expected an indented block` | O corpo da função não está recuado | Coloque 4 espaços (ou Tab) no começo das linhas de dentro |
| `NameError: name 'somar' is not defined` | Chamou a função antes de criar, ou escreveu o nome diferente | Rode a célula do `def` antes. Confira maiúsculas e minúsculas |
| `TypeError: somar() missing 1 required positional argument` | Faltou argumento na chamada | `somar(2, 3)`, não `somar(2)` |
| `TypeError: somar() takes 2 positional arguments but 3 were given` | Passou argumento a mais | Confira quantos parâmetros a função tem |
| A variável fica `None` | A função usa `print` em vez de `return` | Troque por `return` (seção 8) |
| Aparece `<function ...>` em vez do resultado | Esqueceu os parênteses na chamada | `dizer_oi()`, não `dizer_oi` |

Rode a célula abaixo para ver um desses erros de verdade, sem medo:

````python
try:
    somar(2)            # falta um argumento
except TypeError as erro:
    print("Erro:", erro)
````

## 14. Colinha

```python

def nome_da_funcao(param1, param2="padrão"):

    return resultado
```
