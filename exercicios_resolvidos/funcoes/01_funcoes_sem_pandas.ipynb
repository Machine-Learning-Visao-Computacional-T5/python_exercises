## Demo — Definir, chamar e guardar o retorno

`def` cria a função; `saudar("Ana")` a executa. O argumento entra em `nome`; `return` devolve o resultado. `print` só mostra algo na tela.

```python
def saudar(nome):
    return f"Olá, {nome}!"

mensagem = saudar("Ana")
print(mensagem)

def mostrar(nome):
    print(f"Olá, {nome}!")

retorno = mostrar("Ana")
print(retorno)  # None: não houve return
```

## Demo — Antes e depois

Uma regra repetida exige alterações em vários lugares. A função centraliza a regra. Ela também deve ser chamada: definir não calcula os resultados.

```python
antes = [100 * 0.9, 200 * 0.9, 300 * 0.9]
def descontar(preco, percentual=10):
    return preco * (1 - percentual / 100)

depois = [descontar(100), descontar(200), descontar(300)]

print(depois)
print(descontar(preco=100, percentual=20))
```

## Demo — Função como argumento e lambda

`transformar` recebe uma função e chama essa função para cada item. Passe `dobrar`, sem parênteses; `dobrar(3)` executa imediatamente e devolve um número. Lambda cria uma função curta de uma expressão; use def para regras com várias etapas.

```python
def dobrar(numero):
    return numero * 2

resultado = dobrar(3)
print(resultado)  # 6
```

```python
def transformar(valores, operacao):
    resultados = []

    for valor in valores:
        resultado = operacao(valor)
        resultados.append(resultado)

    return resultados
```

```python
print(transformar([1, 2, 3], dobrar))
```

Passamos dobrar sem parênteses, porque estamos entregando a função para transformar usar.
Dentro de transformar, o parâmetro operacao recebe essa função. Assim, operacao(valor) faz:
Chamada	Equivale a	Resultado
operacao(1)	dobrar(1)	2
operacao(2)	dobrar(2)	4
operacao(3)	dobrar(3)	6

```python
print(transformar([1, 2, 3], lambda numero: numero * 2))
```

Parte	Significado
lambda	Cria uma função
numero	Parâmetro que recebe o valor
numero * 2	Expressão cujo resultado será devolvido


Na lambda, não escrevemos return: o resultado da expressão é devolvido automaticamente.

### Solução 5

```python
def transformar(valores, operacao):
    resultados = []

    for valor in valores:
        resultados.append(operacao(valor))

    return resultados


def triplicar(numero):
    return numero * 3


# Usando uma função com nome
a = transformar([1, 2, 3], triplicar)

# Usando uma função lambda
b = transformar([1, 2, 3], lambda numero: numero * 3)

print(a)  # [3, 6, 9]
print(b)  # [3, 6, 9]

# Mostra se os resultados são iguais
print(a == b)  # True
```
