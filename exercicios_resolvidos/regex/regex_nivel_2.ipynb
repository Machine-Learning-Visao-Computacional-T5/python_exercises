### Aplicar a uma coluna — 49–55 minutos
“No módulo `re`, entreguei uma string. Com `.str`, aplico operações de texto à coluna.”

- `produtos["descricao"]` seleciona a coluna; `.str` dá acesso às operações de texto.
- `contains("CAM-", regex=False, na=False)` busca o texto literal e trata ausência como False. Isso não valida o código.
- `[A-Z]{3}` pede três letras maiúsculas; `\d{3}` pede três dígitos.
- `(?P<codigo>...)` cria um grupo nomeado para a coluna `codigo`; `(?P<tamanho>...)` cria outro para `tamanho`.
- `P|M|G` significa uma das três alternativas.
- `extract` devolve os trechos capturados em colunas; entradas sem correspondência aparecem como ausentes.
- `fullmatch(..., na=False)` verifica o formato completo de cada descrição.

Não entre em toda a sintaxe de grupos hoje: mostre a relação entre grupo e coluna. As três primeiras descrições são válidas. “CAM-12 tamanho M”, texto livre e ausência são inválidos para esta regra. Para extrair somente registros integralmente válidos, use a máscara `fullmatch` como no gabarito 1.

**Pergunta:** encontrar “CAM-” comprova que a descrição está correta? **Não**. A busca literal e a validação completa respondem a perguntas diferentes.

## Demo Identificação e extração
\d representa dígito; {3} exige três ocorrências; ^ e $ delimitam começo e fim. Parênteses criam grupos de extração; fullmatch exige o texto inteiro.

```python
produtos = pd.DataFrame({"descricao": ["CAM-001 tamanho M","CAL-023 tamanho G","CAM-104 tamanho P","CAM-12 tamanho M","texto livre",None]})
produtos 

```

```python
print(produtos["descricao"].str.contains("CAM-", regex=False, na=False))
```

```python
padrao = r"(?P<codigo>[A-Z]{3}-\d{3}) tamanho (?P<tamanho>P|M|G)"
padrao
extraidos = produtos["descricao"].str.extract(padrao)
extraidos
```

```python
display(produtos["descricao"].str.fullmatch(padrao, na=False))
```

```python
produtos = pd.DataFrame({"descricao": ["CAM-001 tamanho M","CAL-023 tamanho G","CAM-104 tamanho P","CAM-12 tamanho M","texto livre",None]})
print(produtos["descricao"].str.contains("CAM-", regex=False, na=False))
padrao = r"(?P<codigo>[A-Z]{3}-\d{3}) tamanho (?P<tamanho>P|M|G)"
extraidos = produtos["descricao"].str.extract(padrao)
display(extraidos)
display(produtos["descricao"].str.fullmatch(padrao, na=False))
```

Crie extrair_codigo(texto) usando re.fullmatch. Retorne o código ou None para nulos e formatos inválidos. Passe a função para apply.

```python
import re
descricoes = pd.Series(["CAM-001 tamanho M","CAM-12 tamanho M",None])
# Sua solução aqui.
```

### Resolução para mostrar na aula

### 2 Função por texto e apply

Há várias etapas e tratamento de ausência, por isso def é mais claro que uma lambda longa.

Execute a célula abaixo depois das entradas do exercício. Leia os comentários e compare o resultado com a regra pedida. Os `assert` verificam o resultado esperado; quando passam, não mostram mensagem.

```python
def extrair_codigo(texto):
    if pd.isna(texto):
        return None
    match = re.fullmatch(r"([A-Z]{3}-\d{3}) tamanho (P|M|G)",texto)
    return match.group(1) if match else None

codigos = descricoes.apply(extrair_codigo)
print(codigos.tolist())
assert codigos.iloc[0] == "CAM-001"
assert codigos.iloc[1:].isna().all()
```
