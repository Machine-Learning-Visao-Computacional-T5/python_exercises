### O que é regex? — 27–32 minutos
“Regex é uma forma de descrever um padrão de texto. Uso quando quero procurar, extrair, validar um formato ou substituir partes que seguem uma regra.” Para uma palavra fixa, uma busca simples pode bastar.

| Padrão | Como ler |
|---|---|
| `CAM-` | Estas letras e o hífen literalmente |
| `[0-9]` | Um dígito de 0 a 9 |
| `{3}` | Exatamente três repetições do item anterior |
| `\d` | Um dígito; em strings Python também aceita outros dígitos Unicode |
| `\D` | Um caractere que não é dígito |
| `\s+` | Um ou mais caracteres de espaço, tabulação ou quebra de linha |
| `\b` | Fronteira entre caractere de palavra e não palavra |

`r"..."` é uma string raw: facilita escrever barras sem que Python as interprete como escapes comuns. O `r` não é a regra de regex; a regra está dentro das aspas.

Leia `r"CAM-[0-9]{3}"` em voz alta: “CAM, hífen, três dígitos”. Pergunte se `CAM-12` corresponde ao formato completo: **não**.

### Limpar — parte de 32–42 minutos
`import re` carrega o módulo. `re.sub(padrao, substituicao, texto)` substitui ocorrências. `r"\D"` encontra não dígitos; `""` os remove. Resultado: **65999991234**. Limpar não comprova que o telefone existe ou tem um tamanho válido.

## Demo — Limpar uma string

re.sub troca ocorrências. No padrão r"\D", D maiúsculo significa caractere que não é dígito. A string raw, prefixo r, facilita escrever barras.

```python
import re
telefone = "(65) 99999-1234"
print(re.sub(r"\D", "", telefone))
```

### Procurar não é validar — parte de 32–42 minutos
`search` procura uma ocorrência dentro do texto. Retorna um objeto de correspondência ou `None`. `group(0)` recupera o trecho encontrado. O `if achado` evita chamar um método de `None`.

`fullmatch` exige que **todo o texto** corresponda. `bool(...)` transforma o resultado em `True` ou `False`. Para “Pedido CAM-123 enviado”, a busca acha **CAM-123**, mas a validação completa dá **False**. Para “CAM-123”, dá **True**.

`match` tenta a partir do início; também não exige consumir todo o texto. `findall` retorna todas as correspondências, útil para a prática posterior. Uma busca simples por três dígitos pode encontrar um trecho dentro de `CAM-1234`; os limites importam.

## Demo — Procurar e validar

search procura uma ocorrência dentro do texto; fullmatch exige que toda a string corresponda. [0-9] representa dígitos ASCII e {3} exige três repetições.

```python
texto = "Pedido CAM-123 enviado"
achado = re.search(r"CAM-[0-9]{3}", texto)
print(achado.group(0) if achado else None)
print(bool(re.fullmatch(r"CAM-[0-9]{3}", texto)))
print(bool(re.fullmatch(r"CAM-[0-9]{3}", "CAM-123")))
```

## Demo — Uma regra para vários dados

A função recebe um telefone por chamada. O for chama a mesma regra para cada item. None significa ausência; não vamos transformá-lo na string "None".

```python
def limpar_telefone(texto):
    if texto is None:
        return None
    return re.sub(r"\D", "", texto)

for telefone in ["(65) 99999-1234", "(11) 3333-4444", None]:
    print(limpar_telefone(telefone))
```
