## Demo: função, chamada e variável

Em `def calcular_dobro(numero)`, `numero` é o parâmetro. Em `calcular_dobro(7)`, 7 é o argumento. A chamada executa o corpo e `return` devolve 14. A atribuição guarda esse retorno em `resultado`.

`calcular_dobro` sem parênteses é a própria função. `calcular_dobro(7)` é uma chamada. A variável que recebe o resultado pode ter outro nome.

```python
def calcular_dobro(numero):
    return numero * 2

resultado = calcular_dobro(7)
print(resultado)
valor_final = resultado + 3
print(valor_final)

operacao = calcular_dobro  # guarda a função, sem executá-la
outro_resultado = operacao(5)  # chama a função por esse nome
print(outro_resultado)
```

## Demo: print e return

`print` mostra na tela. `return` entrega um valor para quem chamou a função. Uma função sem return explícito devolve None. Para reutilizar um cálculo, retorne o valor e imprima fora da função.

```python
def apenas_mostrar(numero):
    print(numero * 2)

def devolver_dobro(numero):
    return numero * 2

sem_resultado = apenas_mostrar(4)
com_resultado = devolver_dobro(4)
print(sem_resultado)  # None
print(com_resultado + 1)  # 9
```

## Demo: lambda também é uma função

`lambda x: x + 1` cria uma função com parâmetro x e uma expressão de resultado, sem escrever return. Aqui ela é passada a uma função que vai chamá-la. Use lambda para uma regra curta; use def para funções nomeadas, validação ou várias etapas. Para funções nomeadas, prefira def em vez de atribuir uma lambda a uma variável.

```python
def aplicar(valor, funcao):
    return funcao(valor)

resultado_lambda = aplicar(5, lambda x: x + 1)
print(resultado_lambda)

nomes_ordenados = sorted(["Bruno", "ana", "Carla"], key=lambda nome: nome.lower())
print(nomes_ordenados)
```

## Demo: main organiza a execução

`main` é um nome convencional, não uma palavra especial que executa sozinha. No notebook, defina e depois chame `main()`. Aqui main devolve um valor, que também pode ser guardado. A variável criada dentro de main é local à função.

No arquivo `.py`, `if __name__ == "__main__":` chama main quando o arquivo é executado diretamente. Ao importar esse arquivo, o bloco não roda. A função main não é obrigatória; ajuda a organizar a sequência do programa.

```python
def calcular_subtotal(preco, quantidade):
    return preco * quantidade

def main():
    subtotal = calcular_subtotal(20, 3)
    return subtotal

resultado_programa = main()
print(resultado_programa)
```

### Solução 1 — Chamar e guardar

### Solução 5 — Argumentos nomeados

### Solução 7 — Função e for

Há uma chamada por item; append guarda cada retorno.

```python
def quadrado(numero):
    return numero ** 2
quadrados = []
for numero in numeros:
    resultado = quadrado(numero)
    quadrados.append(resultado)
assert quadrados == [4, 9, 16]
```

### Solução 8 — Validar antes de calcular

A função concentra a validação e a regra.

```python
def dividir(a, b):
    if b == 0:
        raise ValueError("Divisor não pode ser zero.")
    return a / b
resultado = dividir(10, 2)
assert resultado == 5
try:
    dividir(10, 0)
except ValueError as erro:
    print(erro)
else:
    raise AssertionError("Faltou validar divisor zero")
```

### Solução 9 — Vários valores retornados

A função retorna uma tupla; a atribuição desempacota seus itens.

```python
def extremos(valores):
    if not valores:
        raise ValueError("Lista vazia.")
    return min(valores), max(valores)
menor, maior = extremos([8, 2, 5])
assert (menor, maior) == (2, 8)
```

### Solução 20 — Quando trocar lambda por def

Use def para a versão com validação e várias etapas.

```python
classe = aplicar(8, lambda nota: "aprovado" if nota >= 7 else "revisão")
def classificar_nota(nota):
    if not 0 <= nota <= 10:
        raise ValueError("Nota fora do intervalo.")
    return "aprovado" if nota >= 7 else "revisão"
assert classe == classificar_nota(8)
try:
    classificar_nota(11)
except ValueError as erro:
    print(erro)
else:
    raise AssertionError("Nota inválida foi aceita")
```

### Solução 21 — Primeira main

main organiza chamadas; é preciso chamá-la para executar.

```python
def resumo_compra(preco, quantidade):
    subtotal = calcular_subtotal(preco, quantidade)
    total = com_desconto(subtotal)
    return {"subtotal": subtotal, "total": total}
def main():
    resumo = resumo_compra(20, 3)
    return resumo
resultado_main = main()
assert resultado_main == {"subtotal": 60, "total": 54}
```

### Solução 22 — Main com lista

O retorno final reúne os resultados das chamadas individuais.

```python
def main_varios():
    resultados = []
    for pedido in pedidos:
        resumo = resumo_compra(pedido["preco"], pedido["quantidade"])
        resultados.append(resumo)
    return resultados
resultado_varios = main_varios()
assert [r["total"] for r in resultado_varios] == [18, 54]
```

### Solução 23 — Main recebe dados

Receber os dados facilita reutilizar a função com outras listas.

```python
def main_com_dados(pedidos):
    resumos = [resumo_compra(p["preco"], p["quantidade"]) for p in pedidos]
    return sorted(resumos, key=lambda resumo: resumo["total"])
resultado_dados = main_com_dados(pedidos)
assert [r["total"] for r in resultado_dados] == [18, 54]
assert main_com_dados([]) == []
```
