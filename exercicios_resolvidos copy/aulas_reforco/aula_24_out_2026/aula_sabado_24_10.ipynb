## 1. Git e GitHub em uma frase

**Git** é um programa no seu computador que tira "fotos" do seu projeto a cada etapa. **GitHub** é um site que guarda essas fotos na nuvem para você não perder nada e poder trabalhar com outras pessoas.

Pense num videogame com pontos de salvamento:

- **Git** é o botão de "salvar jogo". Cada save guarda o estado exato do projeto, e você pode voltar para qualquer save antigo.
- **GitHub** é o "save na nuvem". Se o seu notebook quebrar, os saves continuam lá, e seus amigos podem baixar e jogar junto.

Detalhe importante: **Git funciona sem GitHub** (só no seu PC), mas o GitHub precisa do Git. Existem outros sites parecidos com o GitHub (GitLab, Bitbucket), e tudo o que você aprender aqui vale para eles também.

**Por que usar?**

- Nunca mais ter `projeto_final_v2_AGORA_VAI.zip`.
- Ver exatamente o que mudou, quando e por quem.
- Desfazer erros voltando para uma versão que funcionava.
- Trabalhar em equipe sem um sobrescrever o arquivo do outro.
- Ter um portfólio público de projetos (recrutadores olham o GitHub).

## 2. Dicionário: as palavras que você vai ouvir

| Palavra | O que é | Na vida real |

## 4. Terminal sem medo

O terminal é só um jeito de conversar com o computador digitando, em vez de clicar. Você digita um comando, aperta Enter, ele responde.

**Como abrir o terminal dentro do VS Code:** menu **Terminal → New Terminal**, ou o atalho **Ctrl + '** (aspa simples, ao lado do 1). Ele aparece na parte de baixo da tela, já dentro da pasta do projeto.

**Lendo o que aparece na tela.** A linha antes do cursor se chama *prompt* e mostra onde você está:

```text

```

- `~/Documentos/meu-site` é a pasta atual (o `~` significa "minha pasta de usuário").
- `(main)` aparece quando a pasta é um repositório Git e mostra a branch atual.
- `$` é onde você digita. Não copie o `$` quando ver comandos em tutoriais.

**Comandos de navegação (servem para tudo, não só Git):**

| Comando | O que faz | Exemplo |
| --- | --- | --- |

| `ls` | Lista os arquivos da pasta (no PowerShell/CMD: `dir`) | `ls` |

## 5. O ciclo do Git: os 4 lugares onde seu arquivo pode estar

Todo arquivo passa por quatro lugares, e cada comando leva ele para o próximo. Entender este desenho é entender o Git.



Leia da esquerda para a direita: você edita na pasta, escolhe o que entra com `git add`, salva com `git commit` e envia com `git push`. A seta tracejada é o caminho de volta, quando o GitHub tem algo que seu PC ainda não tem.

**Por que existe o staging?** Para você escolher. Se mexeu em 5 arquivos mas só 2 estão prontos, dá `git add` só nesses 2 e faz um commit limpo com eles.

## 6. Comandos Git essenciais

Com uns 12 comandos você faz 95% do trabalho do dia a dia. Todos começam com `git`.

**Começar um projeto**

| Comando | O que faz |
| --- | --- |
| `git init` | Transforma a pasta atual num repositório (cria a pasta oculta `.git`). Só uma vez por projeto. |
| `git clone URL` | Baixa um repositório do GitHub para o seu PC. Ex.: `git clone https://github.com/usuario/projeto.git` |

**O dia a dia (o ciclo)**

| Comando | O que faz |
| --- | --- |
| `git status` | **O mais importante.** Mostra o que mudou, o que está preparado e em que branch você está. Na dúvida, rode ele. |
| `git add arquivo` | Coloca um arquivo na área de preparação (staging). |
| `git add .` | Coloca **todos** os arquivos modificados no staging. O ponto significa "tudo daqui". |
| `git commit -m "mensagem"` | Cria o save com uma mensagem. Ex.: `git commit -m "Adiciona página de contato"` |
| `git push` | Envia seus commits para o GitHub. |
| `git pull` | Traz as novidades do GitHub para o seu PC. Faça antes de começar a trabalhar. |

**Olhar o histórico**

| Comando | O que faz |
| --- | --- |
| `git log` | Lista os commits (aperte **q** para sair). |
| `git log --oneline` | Mesma coisa, um commit por linha. Bem mais fácil de ler. |
| `git diff` | Mostra linha por linha o que mudou e ainda não foi para o staging. |

**Branches**

| Comando | O que faz |
| --- | --- |
| `git branch` | Lista as branches; a atual tem um `*`. |
| `git switch -c nome` | Cria uma branch nova e já vai para ela. |
| `git switch nome` | Troca para uma branch que já existe. |
| `git merge nome` | Junta a branch `nome` na branch em que você está. |

**Conectar com o GitHub**

| Comando | O que faz |
| --- | --- |
| `git remote add origin URL` | Liga o repositório local a um repositório do GitHub. Só uma vez. |
| `git remote -v` | Mostra a qual endereço o projeto está ligado. |
| `git push -u origin main` | Primeiro push da branch `main`. Depois disso basta `git push`. |

**Como escrever uma boa mensagem de commit:** comece com um verbo e diga o que a mudança faz. Bom: `Corrige erro no botão de login`. Ruim: `ajustes`, `aaaa`, `agora vai`.

**O que cada pedaço significa** (sim, é chato, mas é aqui que tudo faz sentido):

| Você digita | O que quer dizer | O que acontece |
| --- | --- | --- |
| `git` | "Ei, Git" | Todo comando do Git começa assim, para o terminal saber com quem você está falando |
| `git status` | "Git, como estão as coisas?" | Só **mostra** a situação, não muda nada. Diz em que branch você está e lista os arquivos: vermelho = mudou mas não foi preparado, verde = preparado para o commit. Pode rodar quantas vezes quiser |
| `git add` | "Git, separa isso aqui" | Coloca arquivos na área de preparação (staging). Ainda **não salvou** nada, só escolheu o que vai entrar na foto |
| `.` (o ponto, depois do add) | "tudo desta pasta" | Em vez de escrever o nome de cada arquivo, o ponto pega todos os que mudaram. Para um só: `git add anotacoes.md` |
| `git commit` | "Git, tira a foto" | Salva de verdade tudo o que estava no staging, no histórico do seu PC. Ainda **não foi** para o GitHub |
| `-m` | **m** de **message** (mensagem) | Avisa que a mensagem do commit vem logo em seguida. Todo commit precisa de uma. Se esquecer o `-m`, o Git abre um editor para você escrever (no Vim: digite a mensagem, depois `Esc`, `:wq` e Enter) |
| `"Adiciona anotações da aula"` | A legenda da foto | O texto entre aspas é a mensagem. As aspas são obrigatórias porque a frase tem espaços |
| `git push` | "Git, empurra para a nuvem" | Envia os commits do seu PC para o GitHub. Só depois disso aparece no site |
| `git pull` | "Git, puxa da nuvem" | O contrário do push: traz para o seu PC o que está no GitHub e você ainda não tem |

**Resumindo numa frase:** `status` olha, `add` separa, `commit -m` salva com legenda, `push` envia.

No `git status`, o arquivo aparece vermelho (*untracked*). Depois do `git add`, verde. Depois do commit, *working tree clean*. Depois do `git push`, atualize a página do GitHub: o `anotacoes.md` está lá.

## 8. Git dentro do VS Code (sem digitar comando)

O VS Code faz `add`, `commit` e `push` com cliques. Os números do desenho abaixo correspondem à tabela logo depois dele.



| Nº | Onde fica | Para que serve | Comando equivalente |
| --- | --- | --- | --- |
| 1 | Ícone de três bolinhas ligadas, na barra da esquerda (atalho **Ctrl + Shift + G**) | Abre o painel **Source Control**. O número azul é quantos arquivos mudaram | `git status` |
| 2 | Caixa de texto no topo do painel | Onde você escreve a mensagem do commit | a parte `-m "..."` |
| 3 | Botão **✓ Commit** | Salva o commit. Se nada estiver no staging, o VS Code pergunta se quer incluir tudo: responda **Yes** | `git commit` |
| 4 | Lista **Changes** | Arquivos modificados. O **+** ao lado do nome coloca no staging. A letra mostra o estado: **U** = novo, **M** = modificado, **D** = apagado | `git add arquivo` |
| 5 | Painel **Terminal** embaixo (**Ctrl + '**) | Terminal normal, já dentro da pasta do projeto | qualquer comando |
| 6 | Barra azul no rodapé | Mostra a branch atual (clique para trocar ou criar) e o botão de sincronizar (envia e traz commits) | `git switch` / `git push` + `git pull` |

**Fluxo com cliques:** edite e salve o arquivo → abra o Source Control (1) → clique no **+** do arquivo (4) → escreva a mensagem (2) → clique em **Commit** (3) → clique em **Sync Changes** (o botão vira isso depois do commit) ou no ícone de sincronizar do rodapé (6).

**Clique ou terminal?** Os dois fazem exatamente a mesma coisa. Recomendo aprender pelo terminal primeiro, porque aí você entende o que cada clique está fazendo.

**Dica:** clicar num arquivo da lista Changes abre uma comparação lado a lado, vermelho é o que saiu e verde o que entrou. É o `git diff` com cara bonita.

## 9. Branches e Pull Requests

Branch serve para mexer em algo novo sem arriscar a versão que funciona. Pull Request é como você pede para essa mudança entrar na `main`, com alguém revisando antes.

**Fluxo completo, do jeito que equipes de verdade trabalham:**

1. Atualize a main: `git switch main` e depois `git pull`.
2. Crie uma branch com nome que explica a tarefa: `git switch -c pagina-contato`.
3. Trabalhe normalmente e faça commits: `git add .` e `git commit -m "Cria página de contato"`.
4. Envie a branch: `git push -u origin pagina-contato`.
5. No GitHub vai aparecer uma faixa amarela **"Compare & pull request"**. Clique nela.
6. Escreva um título e uma descrição do que fez e clique em **Create pull request**.
7. Alguém revisa, comenta e aprova. Se pedirem ajustes, faça novos commits na mesma branch e dê `git push`: o PR atualiza sozinho.
8. Clique em **Merge pull request**. A mudança entra na `main`.
9. No seu PC: `git switch main`, `git pull` e pode apagar a branch velha com `git branch -d pagina-contato`.

**Trabalhando sozinho?** Pode fazer merge direto pelo terminal: `git switch main` e `git merge pagina-contato`. Mas fazer PR mesmo sozinho é um ótimo hábito: você revisa seu próprio código antes.

**Contribuindo no projeto de outra pessoa:** clique em **Fork** no repositório dela, clone o *seu* fork, faça a branch e o push nele, e abra o PR do seu fork para o original.

## 10. Deu ruim: erros comuns e como sair

Quase todo erro do Git tem saída. Antes de qualquer coisa, respire e rode `git status`: a resposta costuma dizer o que fazer.

| Mensagem ou situação | O que significa | Como resolver |
| --- | --- | --- |
| `fatal: not a git repository` | Você está numa pasta que não é repositório | Use `pwd` para ver onde está e `cd` até a pasta certa, ou rode `git init` |
| `nothing to commit, working tree clean` | Não é erro: está tudo salvo | Nada a fazer |
| `Please tell me who you are` | Você pulou a configuração de nome e e-mail | Rode os dois `git config --global` da seção 3 |
| `rejected ... fetch first` no push | O GitHub tem commits que você ainda não tem | `git pull` e depois `git push` de novo |
| `CONFLICT (content): Merge conflict in arquivo` | Você e outra pessoa mudaram a mesma linha | Abra o arquivo no VS Code, escolha **Accept Current**, **Accept Incoming** ou **Accept Both**, salve, `git add .` e `git commit` |
| `error: src refspec main does not match any` | Tentou dar push antes do primeiro commit | Faça `git add .` e `git commit` primeiro |
| `remote origin already exists` | Já ligou o repositório antes | Confira com `git remote -v`; para trocar: `git remote set-url origin URL-NOVA` |
| Errei a mensagem do último commit | Ainda não deu push | `git commit --amend -m "Mensagem certa"` |
| Fiz `git add` num arquivo sem querer | Ainda não commitou | `git restore --staged arquivo` |
| Quero jogar fora as mudanças de um arquivo | Voltar ao último commit | `git restore arquivo` (**cuidado:** apaga suas mudanças de vez) |
| Commitei uma senha ou arquivo gigante | Ele fica no histórico | Troque a senha imediatamente e adicione o arquivo ao `.gitignore`. Apagar do histórico é avançado: peça ajuda |

**Os comandos perigosos.** Fuja destes até entender bem: `git reset --hard`, `git push --force` e `git clean -fd`. Eles apagam trabalho sem perguntar.

### Exercício 8: trabalhar numa branch

**Objetivo:** fazer uma mudança separada e depois juntar na main.

1. Crie e entre na branch: `git switch -c lista-compras`.
2. Rode `git branch`. **Repare:** o `*` está em `lista-compras`.
3. Crie `compras.md` com 3 itens. Faça `git add .` e `git commit -m "Cria lista de compras"`.
4. Volte para a main: `git switch main`. **Repare:** o `compras.md` sumiu da pasta! Ele está guardado na outra branch.
5. Junte: `git merge lista-compras`. O arquivo volta.
6. Envie: `git push`.

**Deu certo se:** o `compras.md` aparece no GitHub na branch `main`.

### Exercício 9 (desafio): Pull Request

**Objetivo:** fazer o fluxo que as equipes usam.

1. `git switch -c melhora-readme`
2. No `README.md`, escreva uma descrição do projeto. Commit com mensagem que explique a mudança.
3. `git push -u origin melhora-readme`
4. No GitHub, clique na faixa amarela **Compare & pull request**, escreva uma descrição e clique em **Create pull request**.
5. Revise as mudanças na aba **Files changed** e clique em **Merge pull request** → **Confirm merge**.
6. No PC: `git switch main` e `git pull`.

**Deu certo se:** o README novo aparece no GitHub e no seu PC, e a aba **Pull requests** mostra o seu PR como **Merged** (roxo).

### Exercício 10 (desafio): resolver um conflito

**Objetivo:** perder o medo de conflito.

1. No site do GitHub, edite a primeira linha do `sobre-mim.md` para `Versão do site` e faça o commit.
2. No PC, **sem dar pull**, mude a mesma primeira linha para `Versão do PC`. Faça `git add .` e `git commit -m "Muda primeira linha no PC"`.
3. Rode `git push`. Vai dar erro (*rejected*). É esperado!
4. Rode `git pull`. O Git avisa *CONFLICT*.
5. Abra o arquivo no VS Code: ele mostra as duas versões. Clique em **Accept Current**, **Accept Incoming** ou **Accept Both**.
6. Salve, depois `git add .`, `git commit -m "Resolve conflito"` e `git push`.

**Deu certo se:** o push funcionou e o arquivo no GitHub tem a versão que você escolheu.

### Perguntas para fixar

1. Qual a diferença entre `git commit` e `git push`?
2. Para que serve o `-m`?
3. Se o `git status` mostra um arquivo em vermelho, qual comando você roda para deixar ele verde?
4. Você chegou num computador novo para continuar o projeto. Qual comando roda primeiro: `git clone` ou `git pull`? E no dia seguinte, no mesmo computador?
5. O que o `.` significa em `git add .`?

**Respostas:** (1) commit salva no PC, push envia para o GitHub. (2) diz que a mensagem do commit vem em seguida. (3) `git add` + nome do arquivo (ou `git add .`). (4) no computador novo, `git clone`; no dia seguinte, `git pull`. (5) todos os arquivos modificados da pasta atual.

Exercício
Exercício: criar um arquivo em uma branch e enviá-lo ao GitHub

Você vai usar um repositório já clonado no computador, no qual tenha permissão para enviar alterações. No VS Code, abra a pasta desse projeto e depois Terminal → Novo Terminal.

Execute um comando de cada vez.

1. Confira a situação do projeto
git status

Mostra a branch atual e se existem alterações pendentes. Para começar este exercício, o ideal é aparecer working tree clean, indicando que não há alterações pendentes.

2. Vá para a branch principal
git checkout main

Muda para a branch main, que será a base do exercício. Se o seu repositório usa master, substitua main por master.

3. Atualize o projeto
git pull

Busca e incorpora as atualizações do repositório remoto na branch atual.

4. Crie sua branch

O formato do comando é:
git checkout -b <nome-da-branch>

Substitua <nome-da-branch> pelo nome escolhido, sem os sinais < >. Por exemplo:
git checkout -b exercicio-git-hellen

O -b cria uma branch nova, e o checkout já muda para ela. Uma branch permite desenvolver uma alteração separadamente da branch principal.

5. Crie um arquivo

No explorador de arquivos do VS Code:
1. Clique em Novo Arquivo.
2. Dê o nome apresentacao-hellen.txt.
3. Escreva:
Meu nome é Hellen.
Estou aprendendo a usar Git.
Esta é minha primeira alteração em uma branch.

Salve com Ctrl + S ou Cmd + S, no Mac.

6. Confira o que mudou
git status

O arquivo deve aparecer como untracked, ou seja: ele existe na pasta, mas ainda não está sendo acompanhado pelo Git.

7. Prepare o arquivo para o commit
git add apresentacao-hellen.txt

Seleciona esse arquivo para entrar no próximo commit.

8. Registre a alteração
git commit -m "Adiciona arquivo de apresentação"

Cria um registro da alteração no histórico local do projeto. A opção -m permite escrever uma mensagem explicando o que foi feito.

9. Envie sua branch ao GitHub
git push -u origin exercicio-git-hellen

Envia os commits da sua branch ao repositório remoto:
- origin: nome usual do repositório remoto.
- exercicio-git-hellen: branch que você está enviando.
- -u: associa a branch local à remota. Nos próximos envios dessa branch, você poderá usar apenas git push.
10. Abra um Pull Request

No GitHub, abra o repositório e clique em Compare & pull request. Confira:
- Base: main.
- Compare: exercicio-git-hellen.
Use o título “Adiciona arquivo de apresentação” e clique em Create pull request.

O Pull Request é um pedido para revisar e incorporar suas alterações à branch principal. Abri-lo não faz a mesclagem automaticamente.

Para praticar mais: acrescente uma frase ao arquivo, salve e execute:
git diff
git add apresentacao-hellen.txt
git commit -m "Atualiza apresentação"
git push

O git diff mostra as alterações ainda não preparadas com git add. Depois do envio, o mesmo Pull Request será atualizado com o novo commit.
